```python
import os
import numpy as np
import cv2
import tensorflow as tf
import warnings

from tkinter import Tk, Label, Button, filedialog, messagebox
from PIL import Image, ImageTk

warnings.filterwarnings("ignore")


# ============================================================
# CONSTANTS
# ============================================================

MODEL_PATH = "Best_Cattle_Breed.h5"
DATA_DIR = "data"


# ============================================================
# CHECK FILES
# ============================================================

if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError(
        f"Model file not found: {MODEL_PATH}\n"
        "Make sure Best_Cattle_Breed.h5 is in the same folder as this notebook."
    )

if not os.path.exists(DATA_DIR):
    raise FileNotFoundError(
        f"Data folder not found: {DATA_DIR}\n"
        "Make sure the data folder is in the same folder as this notebook."
    )


# ============================================================
# LOAD MODEL
# ============================================================

print("Loading model...")

model = tf.keras.models.load_model(MODEL_PATH)

print("Model loaded successfully!")


# ============================================================
# LOAD CLASS NAMES
# ============================================================

CLASS_NAMES = sorted(
    [
        folder
        for folder in os.listdir(DATA_DIR)
        if os.path.isdir(os.path.join(DATA_DIR, folder))
    ]
)

if len(CLASS_NAMES) == 0:
    raise ValueError(
        "No class folders were found inside the data folder."
    )

print("Classes found:")
for i, class_name in enumerate(CLASS_NAMES):
    print(i, ":", class_name)

print("Total classes:", len(CLASS_NAMES))


# ============================================================
# IMAGE PREPROCESSING
# ============================================================

def preprocess_image(image_path):

    # Read image
    img = cv2.imread(image_path)

    if img is None:
        raise ValueError("Could not read the selected image.")

    # OpenCV reads images as BGR.
    # Convert BGR -> RGB.
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Resize to model input size
    img = cv2.resize(img, (224, 224))

    # Convert to float32
    img = img.astype(np.float32)

    # EfficientNetV2 preprocessing
    img = tf.keras.applications.efficientnet_v2.preprocess_input(img)

    # Add batch dimension
    # (224,224,3) -> (1,224,224,3)
    img = np.expand_dims(img, axis=0)

    return img


# ============================================================
# PREDICTION FUNCTION
# ============================================================

def predict_image(image_path):

    try:

        # Preprocess image
        img = preprocess_image(image_path)

        # Make prediction
        predictions = model.predict(img, verbose=0)

        # Get prediction probabilities
        probabilities = predictions[0]

        # Get class with highest probability
        class_id = np.argmax(probabilities)

        # Check class ID
        if class_id >= len(CLASS_NAMES):
            raise ValueError(
                "Model output classes do not match the classes "
                "inside the data folder."
            )

        # Get class name
        predicted_class = CLASS_NAMES[class_id]

        # Get confidence
        confidence = float(probabilities[class_id]) * 100

        return predicted_class, confidence

    except Exception as e:

        messagebox.showerror(
            "Prediction Error",
            str(e)
        )

        return None, None


# ============================================================
# BROWSE IMAGE FUNCTION
# ============================================================

def browse_image():

    file_path = filedialog.askopenfilename(
        title="Select Cattle Image",
        filetypes=[
            ("Image files", "*.jpg *.jpeg *.png *.bmp"),
            ("All files", "*.*")
        ]
    )

    # If user cancels
    if not file_path:
        return

    try:

        # Open image using PIL
        image = Image.open(file_path)

        # Display image
        image = image.resize((300, 300))

        photo = ImageTk.PhotoImage(image)

        image_label.config(image=photo)

        # Keep reference to image
        image_label.image = photo

        # Make prediction
        predicted_class, confidence = predict_image(file_path)

        if predicted_class is not None:

            result_label.config(
                text=(
                    f"Prediction: {predicted_class}\n"
                    f"Confidence: {confidence:.2f}%"
                )
            )

    except Exception as e:

        messagebox.showerror(
            "Error",
            str(e)
        )


# ============================================================
# GUI SETUP
# ============================================================

root = Tk()

root.title("Cattle Breed Classifier")

root.geometry("400x550")

root.resizable(False, False)


# ============================================================
# TITLE
# ============================================================

title_label = Label(
    root,
    text="Indian Cattle Breed Classifier",
    font=("Arial", 16, "bold")
)

title_label.pack(pady=15)


# ============================================================
# IMAGE DISPLAY
# ============================================================

image_label = Label(
    root,
    text="No image selected",
    width=35,
    height=15
)

image_label.pack(pady=10)


# ============================================================
# SELECT IMAGE BUTTON
# ============================================================

browse_btn = Button(
    root,
    text="Select Cattle Image",
    font=("Arial", 12),
    command=browse_image
)

browse_btn.pack(pady=20)


# ============================================================
# RESULT LABEL
# ============================================================

result_label = Label(
    root,
    text="",
    font=("Arial", 14)
)

result_label.pack(pady=10)


# ============================================================
# START GUI
# ============================================================

root.mainloop()
```

```python
import os
import numpy as np
import cv2
import tensorflow as tf
import tkinter as tk
from tkinter import filedialog, messagebox

# =========================================================
# SIH CATTLE BREED CLASSIFICATION SYSTEM
# =========================================================

MODEL_PATH = "Best_Cattle_Breed (1).h5"

CLASS_NAMES = [
    "Alambadi", "Amritmahal", "Ayrshire", "Banni", "Bargur",
    "Bhadawari", "Brown_Swiss", "Dangi", "Deoni", "Gir",
    "Guernsey", "Hallikar", "Hariana", "Holstein_Friesian",
    "Jaffrabadi", "Jersey", "Kangayam", "Kankrej", "Kasargod",
    "Kenkatha", "Kherigarh", "Khillari", "Krishna_Valley",
    "Malnad_gidda", "Mehsana", "Murrah", "Nagori", "Nagpuri",
    "Nili_Ravi", "Nimari", "Ongole", "Pulikulam", "Rathi",
    "Red_Dane", "Red_Sindhi", "Sahiwal", "Surti", "Tharparkar",
    "Toda", "Umblachery", "Vechur"
]

print("======================================")
print("   SIH CATTLE BREED CLASSIFIER")
print("======================================")

# ---------------------------------------------------------
# LOAD MODEL
# ---------------------------------------------------------

print("\nLoading model...")

if not os.path.exists(MODEL_PATH):
    print("ERROR: Model not found!")
    print("Looking for:", MODEL_PATH)
    input("\nPress Enter to exit...")
    exit()

model = tf.keras.models.load_model(MODEL_PATH)

print("Model loaded successfully!")
print("Model outputs:", model.output_shape[-1])
print("Number of classes:", len(CLASS_NAMES))

# ---------------------------------------------------------
# PREDICTION
# ---------------------------------------------------------

def predict_image(image_path):

    image = cv2.imread(image_path)

    if image is None:
        messagebox.showerror(
            "Error",
            "Unable to read image."
        )
        return

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    image = cv2.resize(
        image,
        (224, 224)
    )

    image = image.astype(
        np.float32
    )

    image = np.expand_dims(
        image,
        axis=0
    )

    image = tf.keras.applications.efficientnet_v2.preprocess_input(
        image
    )

    prediction = model.predict(
        image,
        verbose=0
    )

    probabilities = prediction[0]

    index = np.argmax(probabilities)

    breed = CLASS_NAMES[index]

    confidence = probabilities[index] * 100

    result = (
        "Predicted Breed: " + breed +
        "\n\nConfidence: " +
        f"{confidence:.2f}%"
    )

    print("\n======================================")
    print("PREDICTION")
    print("======================================")
    print("Breed:", breed)
    print("Confidence:", f"{confidence:.2f}%")

    messagebox.showinfo(
        "Cattle Breed Prediction",
        result
    )


# ---------------------------------------------------------
# SELECT IMAGE
# ---------------------------------------------------------

def select_image():

    file_path = filedialog.askopenfilename(
        title="Select Cattle Image",
        filetypes=[
            ("Image Files",
             "*.jpg *.jpeg *.png *.bmp"),
            ("All Files", "*.*")
        ]
    )

    if file_path:

        print("\nSelected image:")
        print(file_path)

        predict_image(
            file_path
        )


# ---------------------------------------------------------
# UI
# ---------------------------------------------------------

root = tk.Tk()

root.title(
    "SIH Cattle Breed Classification"
)

root.geometry(
    "600x400"
)

title = tk.Label(
    root,
    text="🐄 Cattle Breed Classification",
    font=("Arial", 24, "bold")
)

title.pack(
    pady=50
)

info = tk.Label(
    root,
    text="AI-based classification of 41 cattle breeds",
    font=("Arial", 13)
)

info.pack(
    pady=10
)

button = tk.Button(
    root,
    text="SELECT CATTLE IMAGE",
    font=("Arial", 15, "bold"),
    padx=30,
    pady=15,
    command=select_image
)

button.pack(
    pady=30
)

footer = tk.Label(
    root,
    text="Smart India Hackathon Project",
    font=("Arial", 10)
)

footer.pack(
    side="bottom",
    pady=20
)

# ---------------------------------------------------------
# START
# ---------------------------------------------------------

print("\nStarting application...")

root.mainloop()
```
