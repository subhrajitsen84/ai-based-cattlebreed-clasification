```python
import os
import warnings
import numpy as np
import cv2
import tensorflow as tf

from tkinter import Tk, Label, Button, filedialog, messagebox, Frame
from PIL import Image, ImageTk

warnings.filterwarnings("ignore")

# ============================================================
# SETTINGS
# ============================================================

MODEL_PATH = "Best_Cattle_Breed.h5"
DATA_DIR = "data"

IMAGE_SIZE = (224, 224)

# ============================================================
# LOAD MODEL
# ============================================================

print("Loading model...")

if not os.path.exists(MODEL_PATH):
    print("ERROR: Best_Cattle_Breed.h5 not found!")
    raise FileNotFoundError(
        "Best_Cattle_Breed.h5 was not found in the current folder."
    )

model = tf.keras.models.load_model(MODEL_PATH)

print("Model loaded successfully.")

# ============================================================
# LOAD CLASS NAMES
# ============================================================

if not os.path.exists(DATA_DIR):
    print("ERROR: data folder not found!")
    raise FileNotFoundError(
        "The 'data' folder was not found."
    )

CLASS_NAMES = sorted([
    folder for folder in os.listdir(DATA_DIR)
    if os.path.isdir(os.path.join(DATA_DIR, folder))
])

print("\nClasses detected:")
for i, name in enumerate(CLASS_NAMES):
    print(f"{i} -> {name}")

print(f"\nTotal classes: {len(CLASS_NAMES)}")

# ============================================================
# PREPROCESS IMAGE
# ============================================================

def preprocess_image(image_path):

    image = cv2.imread(image_path)

    if image is None:
        raise ValueError("Could not read the selected image.")

    # OpenCV BGR -> RGB
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Resize for model
    image = cv2.resize(image, IMAGE_SIZE)

    # Convert to float
    image = image.astype(np.float32)

    # EfficientNetV2 preprocessing
    image = tf.keras.applications.efficientnet_v2.preprocess_input(image)

    # Add batch dimension
    image = np.expand_dims(image, axis=0)

    return image


# ============================================================
# PREDICTION
# ============================================================

def predict_breed(image_path):

    processed_image = preprocess_image(image_path)

    prediction = model.predict(processed_image, verbose=0)

    # Handle different model output formats
    if prediction.ndim == 2:
        probabilities = prediction[0]
    else:
        probabilities = prediction.flatten()

    predicted_index = int(np.argmax(probabilities))

    confidence = float(probabilities[predicted_index]) * 100

    if predicted_index < len(CLASS_NAMES):
        breed = CLASS_NAMES[predicted_index]
    else:
        breed = f"Class {predicted_index}"

    return breed, confidence


# ============================================================
# GUI
# ============================================================

root = Tk()

root.title("Cattle Breed Classification System")

# ============================================================
# FULLSCREEN / MAXIMIZED WINDOW
# ============================================================

try:
    root.state("zoomed")
except:
    screen_width = root.winfo_screenwidth()
    screen_height = root.winfo_screenheight()

    root.geometry(
        f"{screen_width}x{screen_height}+0+0"
    )

root.resizable(True, True)

# Minimum size
root.minsize(900, 650)

# ============================================================
# VARIABLES
# ============================================================

selected_image_path = None
display_image = None

# ============================================================
# TITLE
# ============================================================

title_label = Label(
    root,
    text="CATTLE BREED CLASSIFICATION SYSTEM",
    font=("Arial", 26, "bold")
)

title_label.pack(pady=20)

subtitle_label = Label(
    root,
    text="Upload a cattle image to predict its breed",
    font=("Arial", 14)
)

subtitle_label.pack(pady=5)

# ============================================================
# MAIN FRAME
# ============================================================

main_frame = Frame(root)
main_frame.pack(
    fill="both",
    expand=True,
    padx=30,
    pady=20
)

# ============================================================
# IMAGE FRAME
# ============================================================

image_frame = Frame(
    main_frame,
    bd=2,
    relief="groove"
)

image_frame.pack(
    fill="both",
    expand=True,
    padx=20,
    pady=10
)

image_label = Label(
    image_frame,
    text="No image selected",
    font=("Arial", 18)
)

image_label.pack(
    fill="both",
    expand=True,
    padx=20,
    pady=20
)

# ============================================================
# RESULT
# ============================================================

result_label = Label(
    main_frame,
    text="Prediction will appear here",
    font=("Arial", 22, "bold")
)

result_label.pack(pady=15)

confidence_label = Label(
    main_frame,
    text="",
    font=("Arial", 16)
)

confidence_label.pack(pady=5)

# ============================================================
# SHOW IMAGE
# ============================================================

def show_image(image_path):

    global display_image

    try:

        original = Image.open(image_path)

        # Get available area
        image_frame.update_idletasks()

        max_width = image_frame.winfo_width() - 50
        max_height = image_frame.winfo_height() - 50

        # Safety values
        if max_width < 300:
            max_width = 800

        if max_height < 300:
            max_height = 500

        # Keep original aspect ratio
        original.thumbnail(
            (max_width, max_height),
            Image.Resampling.LANCZOS
        )

        display_image = ImageTk.PhotoImage(original)

        image_label.configure(
            image=display_image,
            text=""
        )

        image_label.image = display_image

    except Exception as e:

        messagebox.showerror(
            "Image Error",
            f"Could not display image:\n\n{e}"
        )


# ============================================================
# UPLOAD IMAGE
# ============================================================

def select_image():

    global selected_image_path

    file_path = filedialog.askopenfilename(
        title="Select Cattle Image",
        filetypes=[
            ("Image Files", "*.jpg *.jpeg *.png *.bmp *.webp"),
            ("JPG Files", "*.jpg"),
            ("PNG Files", "*.png"),
            ("All Files", "*.*")
        ]
    )

    if not file_path:
        return

    selected_image_path = file_path

    show_image(file_path)

    result_label.config(
        text="Image selected"
    )

    confidence_label.config(
        text="Click 'Predict Breed'"
    )


# ============================================================
# PREDICT BUTTON
# ============================================================

def predict_selected_image():

    if selected_image_path is None:

        messagebox.showwarning(
            "No Image",
            "Please select a cattle image first."
        )

        return

    try:

        result_label.config(
            text="Predicting..."
        )

        confidence_label.config(
            text=""
        )

        root.update()

        breed, confidence = predict_breed(
            selected_image_path
        )

        result_label.config(
            text=f"Predicted Breed: {breed}"
        )

        confidence_label.config(
            text=f"Confidence: {confidence:.2f}%"
        )

        print(
            f"\nPrediction: {breed}"
        )

        print(
            f"Confidence: {confidence:.2f}%"
        )

    except Exception as e:

        result_label.config(
            text="Prediction failed"
        )

        confidence_label.config(
            text=""
        )

        messagebox.showerror(
            "Prediction Error",
            f"Something went wrong:\n\n{e}"
        )


# ============================================================
# BUTTON FRAME
# ============================================================

button_frame = Frame(root)

button_frame.pack(
    pady=20
)

# ============================================================
# SELECT IMAGE BUTTON
# ============================================================

select_button = Button(
    button_frame,
    text="SELECT CATTLE IMAGE",
    command=select_image,
    font=("Arial", 14, "bold"),
    padx=25,
    pady=12
)

select_button.pack(
    side="left",
    padx=15
)

# ============================================================
# PREDICT BUTTON
# ============================================================

predict_button = Button(
    button_frame,
    text="PREDICT BREED",
    command=predict_selected_image,
    font=("Arial", 14, "bold"),
    padx=25,
    pady=12
)

predict_button.pack(
    side="left",
    padx=15
)

# ============================================================
# EXIT BUTTON
# ============================================================

exit_button = Button(
    button_frame,
    text="EXIT",
    command=root.destroy,
    font=("Arial", 14, "bold"),
    padx=25,
    pady=12
)

exit_button.pack(
    side="left",
    padx=15
)

# ============================================================
# START APPLICATION
# ============================================================

print("\n==========================================")
print(" CATTLE BREED CLASSIFICATION SYSTEM")
print("==========================================")
print("GUI started successfully.")
print("Select an image to begin.")

root.mainloop()
```

```python
import os
import numpy as np
import tensorflow as tf
import cv2
import warnings

from tkinter import (
    Tk,
    Label,
    Button,
    filedialog,
    messagebox,
    Frame
)

from PIL import Image, ImageTk

warnings.filterwarnings("ignore")


# ============================================================
# SETTINGS
# ============================================================

MODEL_PATH = "Best_Cattle_Breed_Improved.h5"

IMAGE_SIZE = (224, 224)

CONFIDENCE_THRESHOLD = 50.0


# ============================================================
# 41 CATTLE BREEDS
# ============================================================

CLASS_NAMES = [
    "Alambadi",
    "Amritmahal",
    "Ayrshire",
    "Banni",
    "Bargur",
    "Bhadawari",
    "Brown_Swiss",
    "Dangi",
    "Deoni",
    "Gir",
    "Guernsey",
    "Hallikar",
    "Hariana",
    "Holstein_Friesian",
    "Jaffrabadi",
    "Jersey",
    "Kangayam",
    "Kankrej",
    "Kasargod",
    "Kenkatha",
    "Kherigarh",
    "Khillari",
    "Krishna_Valley",
    "Malnad_gidda",
    "Mehsana",
    "Murrah",
    "Nagori",
    "Nagpuri",
    "Nili_Ravi",
    "Nimari",
    "Ongole",
    "Pulikulam",
    "Rathi",
    "Red_Dane",
    "Red_Sindhi",
    "Sahiwal",
    "Surti",
    "Tharparkar",
    "Toda",
    "Umblachery",
    "Vechur"
]


# ============================================================
# CHECK MODEL
# ============================================================

print("========================================")
print("CATTLE BREED IDENTIFICATION SYSTEM")
print("========================================")

print("\nChecking model...")

if not os.path.exists(MODEL_PATH):

    print("❌ MODEL NOT FOUND")
    print()
    print("Expected model:")
    print(os.path.abspath(MODEL_PATH))
    print()

    raise FileNotFoundError(
        "Best_Cattle_Breed_Improved.h5 was not found."
    )

print("✅ Model found")
print("Loading model...")


# ============================================================
# LOAD MODEL
# ============================================================

try:

    model = tf.keras.models.load_model(
        MODEL_PATH,
        compile=False
    )

    print("✅ Model loaded successfully")

except Exception as e:

    print("❌ Error loading model:")
    print(e)

    raise


# ============================================================
# CHECK MODEL OUTPUT
# ============================================================

model_output_classes = model.output_shape[-1]

print(
    f"Model output classes: {model_output_classes}"
)

print(
    f"Expected classes: {len(CLASS_NAMES)}"
)

if model_output_classes != len(CLASS_NAMES):

    raise ValueError(
        f"\nModel/class mismatch!\n"
        f"Model outputs: {model_output_classes}\n"
        f"Class names: {len(CLASS_NAMES)}"
    )

print("✅ Model has 41 outputs")
print("✅ Class list matches model")


# ============================================================
# IMAGE PREPROCESSING
# ============================================================

def preprocess_image(image_path):

    image = cv2.imread(image_path)

    if image is None:

        raise ValueError(
            "Unable to read the selected image."
        )

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    image = cv2.resize(
        image,
        IMAGE_SIZE
    )

    image = image.astype(
        np.float32
    )

    image = (
        tf.keras.applications.efficientnet_v2
        .preprocess_input(image)
    )

    image = np.expand_dims(
        image,
        axis=0
    )

    return image


# ============================================================
# PREDICT BREED
# ============================================================

def predict_breed(image_path):

    image = preprocess_image(
        image_path
    )

    predictions = model.predict(
        image,
        verbose=0
    )[0]

    top_indices = np.argsort(
        predictions
    )[-3:][::-1]

    results = []

    for index in top_indices:

        breed = CLASS_NAMES[index]

        confidence = (
            float(predictions[index]) * 100
        )

        results.append(
            (
                breed,
                confidence
            )
        )

    return results


# ============================================================
# DISPLAY IMAGE
# ============================================================

def display_image(image_path):

    global displayed_photo

    image = Image.open(
        image_path
    )

    if image.mode not in (
        "RGB",
        "RGBA"
    ):

        image = image.convert("RGB")

    # Make sure Tkinter has calculated the
    # current fullscreen frame size.

    image_frame.update_idletasks()

    frame_width = image_frame.winfo_width()
    frame_height = image_frame.winfo_height()

    # Leave some padding around image.

    max_width = max(
        300,
        frame_width - 40
    )

    max_height = max(
        250,
        frame_height - 40
    )

    original_width, original_height = image.size

    # Keep original aspect ratio.

    scale = min(
        max_width / original_width,
        max_height / original_height
    )

    new_width = max(
        1,
        int(original_width * scale)
    )

    new_height = max(
        1,
        int(original_height * scale)
    )

    image = image.resize(
        (new_width, new_height),
        Image.Resampling.LANCZOS
    )

    displayed_photo = ImageTk.PhotoImage(
        image
    )

    image_label.config(
        image=displayed_photo,
        text=""
    )

    # Keep reference alive.

    image_label.image = displayed_photo


# ============================================================
# SELECT IMAGE
# ============================================================

def select_image():

    file_path = filedialog.askopenfilename(

        title="Select Cattle Image",

        filetypes=[
            (
                "Image Files",
                "*.jpg *.jpeg *.png *.bmp *.webp"
            ),
            (
                "All Files",
                "*.*"
            )
        ]
    )

    if not file_path:
        return

    try:

        # Show image.

        display_image(
            file_path
        )

        # Update status.

        status_label.config(
            text="Analyzing image..."
        )

        root.update_idletasks()

        # Predict.

        results = predict_breed(
            file_path
        )

        # Best prediction.

        best_breed = results[0][0]

        best_confidence = results[0][1]

        prediction_label.config(
            text=(
                "Predicted Breed:\n"
                + best_breed.replace(
                    "_",
                    " "
                )
            )
        )

        confidence_label.config(
            text=(
                f"Confidence: "
                f"{best_confidence:.2f}%"
            )
        )

        # Top 3.

        top3_text = (
            "Top 3 Predictions\n\n"
        )

        for i, (
            breed,
            confidence
        ) in enumerate(
            results,
            start=1
        ):

            clean_breed = breed.replace(
                "_",
                " "
            )

            top3_text += (
                f"{i}. {clean_breed}"
                f"  —  {confidence:.2f}%\n"
            )

        top3_label.config(
            text=top3_text
        )

        # Confidence warning.

        if best_confidence < CONFIDENCE_THRESHOLD:

            warning_label.config(
                text=(
                    "⚠ LOW CONFIDENCE\n"
                    "Try a clearer cattle image."
                )
            )

        else:

            warning_label.config(
                text=(
                    "✓ Prediction completed"
                )
            )

        status_label.config(
            text="Analysis complete"
        )

    except Exception as e:

        messagebox.showerror(
            "Prediction Error",
            str(e)
        )

        status_label.config(
            text="Prediction failed"
        )


# ============================================================
# RESET
# ============================================================

def reset():

    image_label.config(
        image="",
        text="No image selected"
    )

    image_label.image = None

    prediction_label.config(
        text="Predicted Breed:\n---"
    )

    confidence_label.config(
        text="Confidence: ---"
    )

    top3_label.config(
        text="Top 3 Predictions\n\n---"
    )

    warning_label.config(
        text=""
    )

    status_label.config(
        text="Select an image to begin"
    )


# ============================================================
# CREATE WINDOW
# ============================================================

root = Tk()

root.title(
    "Cattle Breed Identification System"
)


# ============================================================
# MAXIMIZE WINDOW
# ============================================================

# Windows method:
try:

    root.state(
        "zoomed"
    )

except:

    screen_width = root.winfo_screenwidth()
    screen_height = root.winfo_screenheight()

    root.geometry(
        f"{screen_width}x{screen_height}+0+0"
    )


# Allow resizing.

root.resizable(
    True,
    True
)

# Minimum size.

root.minsize(
    900,
    650
)


# ============================================================
# TITLE
# ============================================================

title_label = Label(
    root,
    text="🐄 Cattle Breed Identification",
    font=(
        "Arial",
        26,
        "bold"
    )
)

title_label.pack(
    pady=(18, 5)
)


subtitle_label = Label(
    root,
    text=(
        "AI-powered identification of 41 cattle breeds"
    ),
    font=(
        "Arial",
        12
    )
)

subtitle_label.pack(
    pady=(0, 12)
)


# ============================================================
# IMAGE FRAME
# ============================================================

image_frame = Frame(
    root,
    width=750,
    height=450,
    relief="groove",
    borderwidth=2
)

image_frame.pack(
    pady=5,
    padx=30,
    fill="both",
    expand=True
)

image_frame.pack_propagate(
    False
)


# ============================================================
# IMAGE LABEL
# ============================================================

image_label = Label(
    image_frame,
    text="No image selected",
    font=(
        "Arial",
        16
    ),
    anchor="center"
)

image_label.pack(
    expand=True,
    fill="both"
)


# ============================================================
# BUTTON FRAME
# ============================================================

button_frame = Frame(
    root
)

button_frame.pack(
    pady=12
)


# ============================================================
# SELECT BUTTON
# ============================================================

select_button = Button(
    button_frame,
    text="📷 Select Cattle Image",
    font=(
        "Arial",
        13,
        "bold"
    ),
    command=select_image,
    padx=20,
    pady=8
)

select_button.grid(
    row=0,
    column=0,
    padx=10
)


# ============================================================
# RESET BUTTON
# ============================================================

reset_button = Button(
    button_frame,
    text="🔄 Reset",
    font=(
        "Arial",
        13
    ),
    command=reset,
    padx=25,
    pady=8
)

reset_button.grid(
    row=0,
    column=1,
    padx=10
)


# ============================================================
# PREDICTION
# ============================================================

prediction_label = Label(
    root,
    text="Predicted Breed:\n---",
    font=(
        "Arial",
        19,
        "bold"
    )
)

prediction_label.pack(
    pady=(5, 3)
)


# ============================================================
# CONFIDENCE
# ============================================================

confidence_label = Label(
    root,
    text="Confidence: ---",
    font=(
        "Arial",
        14
    )
)

confidence_label.pack(
    pady=2
)


# ============================================================
# TOP 3
# ============================================================

top3_label = Label(
    root,
    text="Top 3 Predictions\n\n---",
    font=(
        "Arial",
        12
    ),
    justify="left"
)

top3_label.pack(
    pady=5
)


# ============================================================
# WARNING
# ============================================================

warning_label = Label(
    root,
    text="",
    font=(
        "Arial",
        11
    )
)

warning_label.pack(
    pady=3
)


# ============================================================
# STATUS
# ============================================================

status_label = Label(
    root,
    text="Select an image to begin",
    font=(
        "Arial",
        10
    )
)

status_label.pack(
    pady=5
)


# ============================================================
# START
# ============================================================

print()
print("========================================")
print("SYSTEM READY")
print("========================================")
print(
    f"Model: {MODEL_PATH}"
)
print(
    f"Number of breeds: {len(CLASS_NAMES)}"
)
print(
    "Best validation accuracy: 70.31%"
)
print(
    "Waiting for image..."
)
print()

root.mainloop()
```

```python
import os
import numpy as np
import tensorflow as tf
import cv2
import warnings

from tkinter import (
    Tk,
    Label,
    Button,
    filedialog,
    messagebox,
    Frame
)

from PIL import Image, ImageTk

warnings.filterwarnings("ignore")


# ============================================================
# SETTINGS
# ============================================================

MODEL_PATH = "Best_Cattle_Breed_Improved.h5"

IMAGE_SIZE = (224, 224)

CONFIDENCE_THRESHOLD = 50.0


# ============================================================
# 41 CATTLE BREEDS
# ============================================================

CLASS_NAMES = [
    "Alambadi",
    "Amritmahal",
    "Ayrshire",
    "Banni",
    "Bargur",
    "Bhadawari",
    "Brown_Swiss",
    "Dangi",
    "Deoni",
    "Gir",
    "Guernsey",
    "Hallikar",
    "Hariana",
    "Holstein_Friesian",
    "Jaffrabadi",
    "Jersey",
    "Kangayam",
    "Kankrej",
    "Kasargod",
    "Kenkatha",
    "Kherigarh",
    "Khillari",
    "Krishna_Valley",
    "Malnad_gidda",
    "Mehsana",
    "Murrah",
    "Nagori",
    "Nagpuri",
    "Nili_Ravi",
    "Nimari",
    "Ongole",
    "Pulikulam",
    "Rathi",
    "Red_Dane",
    "Red_Sindhi",
    "Sahiwal",
    "Surti",
    "Tharparkar",
    "Toda",
    "Umblachery",
    "Vechur"
]


# ============================================================
# CHECK MODEL
# ============================================================

print("========================================")
print("CATTLE BREED IDENTIFICATION SYSTEM")
print("========================================")

print("\nChecking model...")

if not os.path.exists(MODEL_PATH):

    print("❌ MODEL NOT FOUND")
    print()
    print("Expected model:")
    print(os.path.abspath(MODEL_PATH))
    print()

    raise FileNotFoundError(
        "Best_Cattle_Breed_Improved.h5 was not found."
    )

print("✅ Model found")
print("Loading model...")


# ============================================================
# LOAD MODEL
# ============================================================

try:

    model = tf.keras.models.load_model(
        MODEL_PATH,
        compile=False
    )

    print("✅ Model loaded successfully")

except Exception as e:

    print("❌ Error loading model:")
    print(e)

    raise


# ============================================================
# CHECK MODEL OUTPUT
# ============================================================

model_output_classes = model.output_shape[-1]

print(
    f"Model output classes: {model_output_classes}"
)

print(
    f"Expected classes: {len(CLASS_NAMES)}"
)

if model_output_classes != len(CLASS_NAMES):

    raise ValueError(
        f"\nModel/class mismatch!\n"
        f"Model outputs: {model_output_classes}\n"
        f"Class names: {len(CLASS_NAMES)}"
    )

print("✅ Model has 41 outputs")
print("✅ Class list matches model")


# ============================================================
# IMAGE PREPROCESSING
# ============================================================

def preprocess_image(image_path):

    image = cv2.imread(image_path)

    if image is None:

        raise ValueError(
            "Unable to read the selected image."
        )

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    image = cv2.resize(
        image,
        IMAGE_SIZE
    )

    image = image.astype(
        np.float32
    )

    image = (
        tf.keras.applications.efficientnet_v2
        .preprocess_input(image)
    )

    image = np.expand_dims(
        image,
        axis=0
    )

    return image


# ============================================================
# PREDICT BREED
# ============================================================

def predict_breed(image_path):

    image = preprocess_image(
        image_path
    )

    predictions = model.predict(
        image,
        verbose=0
    )[0]

    # Get TOP 3 indexes
    top_indices = np.argsort(
        predictions
    )[-3:][::-1]

    results = []

    for index in top_indices:

        breed = CLASS_NAMES[index]

        confidence = (
            float(predictions[index]) * 100
        )

        results.append(
            (
                breed,
                confidence
            )
        )

    return results


# ============================================================
# DISPLAY IMAGE
# ============================================================

def display_image(image_path):

    global displayed_photo

    image = Image.open(
        image_path
    )

    if image.mode not in (
        "RGB",
        "RGBA"
    ):

        image = image.convert("RGB")

    # Get actual image frame size

    image_frame.update_idletasks()

    frame_width = image_frame.winfo_width()
    frame_height = image_frame.winfo_height()

    # Leave padding

    max_width = max(
        300,
        frame_width - 40
    )

    max_height = max(
        250,
        frame_height - 40
    )

    original_width, original_height = image.size

    # Maintain aspect ratio

    scale = min(
        max_width / original_width,
        max_height / original_height
    )

    new_width = max(
        1,
        int(original_width * scale)
    )

    new_height = max(
        1,
        int(original_height * scale)
    )

    image = image.resize(
        (
            new_width,
            new_height
        ),
        Image.Resampling.LANCZOS
    )

    displayed_photo = ImageTk.PhotoImage(
        image
    )

    image_label.config(
        image=displayed_photo,
        text=""
    )

    image_label.image = displayed_photo


# ============================================================
# SELECT IMAGE
# ============================================================

def select_image():

    file_path = filedialog.askopenfilename(

        title="Select Cattle Image",

        filetypes=[
            (
                "Image Files",
                "*.jpg *.jpeg *.png *.bmp *.webp"
            ),
            (
                "All Files",
                "*.*"
            )
        ]
    )

    if not file_path:
        return

    try:

        # Show selected image

        display_image(
            file_path
        )

        status_label.config(
            text="Analyzing image..."
        )

        root.update_idletasks()

        # Predict

        results = predict_breed(
            file_path
        )

        # ====================================================
        # BEST PREDICTION
        # ====================================================

        best_breed = results[0][0]

        best_confidence = results[0][1]

        prediction_label.config(
            text=(
                "Predicted Breed: "
                + best_breed.replace(
                    "_",
                    " "
                )
            )
        )

        confidence_label.config(
            text=(
                f"Confidence: "
                f"{best_confidence:.2f}%"
            )
        )

        # ====================================================
        # TOP 3 PREDICTIONS
        # ====================================================

        top3_text = ""

        for i, (
            breed,
            confidence
        ) in enumerate(
            results,
            start=1
        ):

            clean_breed = breed.replace(
                "_",
                " "
            )

            top3_text += (
                f"{i}. {clean_breed} "
                f"— {confidence:.2f}%"
            )

            # New line after each prediction
            if i < 3:
                top3_text += "\n"

        top3_label.config(
            text=top3_text
        )

        # ====================================================
        # CONFIDENCE WARNING
        # ====================================================

        if best_confidence < CONFIDENCE_THRESHOLD:

            warning_label.config(
                text=(
                    "⚠ LOW CONFIDENCE — "
                    "Try a clearer cattle image."
                )
            )

        else:

            warning_label.config(
                text="✓ Prediction completed"
            )

        # ====================================================
        # STATUS
        # ====================================================

        status_label.config(
            text="Analysis complete"
        )

    except Exception as e:

        messagebox.showerror(
            "Prediction Error",
            str(e)
        )

        status_label.config(
            text="Prediction failed"
        )


# ============================================================
# RESET
# ============================================================

def reset():

    image_label.config(
        image="",
        text="No image selected"
    )

    image_label.image = None

    prediction_label.config(
        text="Predicted Breed: ---"
    )

    confidence_label.config(
        text="Confidence: ---"
    )

    top3_label.config(
        text="Top 3 Predictions will appear here"
    )

    warning_label.config(
        text=""
    )

    status_label.config(
        text="Select an image to begin"
    )


# ============================================================
# CREATE WINDOW
# ============================================================

root = Tk()

root.title(
    "Cattle Breed Identification System"
)

# ============================================================
# MAXIMIZE WINDOW
# ============================================================

try:

    root.state("zoomed")

except:

    screen_width = root.winfo_screenwidth()
    screen_height = root.winfo_screenheight()

    root.geometry(
        f"{screen_width}x{screen_height}+0+0"
    )

root.resizable(
    True,
    True
)

root.minsize(
    900,
    650
)


# ============================================================
# TITLE
# ============================================================

title_label = Label(
    root,
    text="🐄 Cattle Breed Identification",
    font=(
        "Arial",
        25,
        "bold"
    )
)

title_label.pack(
    pady=(12, 2)
)


subtitle_label = Label(
    root,
    text="AI-powered identification of 41 cattle breeds",
    font=(
        "Arial",
        12
    )
)

subtitle_label.pack(
    pady=(0, 8)
)


# ============================================================
# IMAGE FRAME
# ============================================================

image_frame = Frame(
    root,
    width=750,
    height=400,
    relief="groove",
    borderwidth=2
)

image_frame.pack(
    pady=5,
    padx=30
)

# IMPORTANT:
# Keep image frame at a fixed height so that
# the prediction section always remains visible.

image_frame.pack_propagate(
    False
)


# ============================================================
# IMAGE LABEL
# ============================================================

image_label = Label(
    image_frame,
    text="No image selected",
    font=(
        "Arial",
        16
    ),
    anchor="center"
)

image_label.pack(
    expand=True,
    fill="both"
)


# ============================================================
# BUTTON FRAME
# ============================================================

button_frame = Frame(
    root
)

button_frame.pack(
    pady=8
)


# ============================================================
# SELECT BUTTON
# ============================================================

select_button = Button(
    button_frame,
    text="📷 Select Cattle Image",
    font=(
        "Arial",
        13,
        "bold"
    ),
    command=select_image,
    padx=20,
    pady=7
)

select_button.grid(
    row=0,
    column=0,
    padx=8
)


# ============================================================
# RESET BUTTON
# ============================================================

reset_button = Button(
    button_frame,
    text="🔄 Reset",
    font=(
        "Arial",
        13
    ),
    command=reset,
    padx=25,
    pady=7
)

reset_button.grid(
    row=0,
    column=1,
    padx=8
)


# ============================================================
# RESULTS FRAME
# ============================================================

results_frame = Frame(
    root
)

results_frame.pack(
    pady=3
)


# ============================================================
# PREDICTED BREED
# ============================================================

prediction_label = Label(
    results_frame,
    text="Predicted Breed: ---",
    font=(
        "Arial",
        18,
        "bold"
    )
)

prediction_label.pack(
    pady=2
)


# ============================================================
# CONFIDENCE
# ============================================================

confidence_label = Label(
    results_frame,
    text="Confidence: ---",
    font=(
        "Arial",
        13
    )
)

confidence_label.pack(
    pady=1
)


# ============================================================
# TOP 3 TITLE
# ============================================================

top3_title = Label(
    results_frame,
    text="Top 3 Predictions",
    font=(
        "Arial",
        12,
        "bold"
    )
)

top3_title.pack(
    pady=(3, 0)
)


# ============================================================
# TOP 3 RESULTS
# ============================================================

top3_label = Label(
    results_frame,
    text="1. ---\n2. ---\n3. ---",
    font=(
        "Arial",
        11
    ),
    justify="center"
)

top3_label.pack(
    pady=1
)


# ============================================================
# WARNING
# ============================================================

warning_label = Label(
    results_frame,
    text="",
    font=(
        "Arial",
        10
    )
)

warning_label.pack(
    pady=1
)


# ============================================================
# STATUS
# ============================================================

status_label = Label(
    results_frame,
    text="Select an image to begin",
    font=(
        "Arial",
        9
    )
)

status_label.pack(
    pady=1
)


# ============================================================
# START APPLICATION
# ============================================================

print()
print("========================================")
print("SYSTEM READY")
print("========================================")
print(
    f"Model: {MODEL_PATH}"
)
print(
    f"Number of breeds: {len(CLASS_NAMES)}"
)
print(
    "Best validation accuracy: 70.31%"
)
print(
    "Waiting for image..."
)
print()

root.mainloop()
```

```python
import os
import numpy as np
import tensorflow as tf
import cv2
from tkinter import Tk, Label, Button, filedialog, messagebox, Frame
from PIL import Image, ImageTk

# ============================================================
# SETTINGS
# ============================================================

MODEL_PATH = "Best_Cattle_Breed_Improved.h5"
IMAGE_SIZE = (224, 224)

CLASS_NAMES = [
    "Alambadi","Amritmahal","Ayrshire","Banni","Bargur","Bhadawari",
    "Brown_Swiss","Dangi","Deoni","Gir","Guernsey","Hallikar","Hariana",
    "Holstein_Friesian","Jaffrabadi","Jersey","Kangayam","Kankrej",
    "Kasargod","Kenkatha","Kherigarh","Khillari","Krishna_Valley",
    "Malnad_gidda","Mehsana","Murrah","Nagori","Nagpuri","Nili_Ravi",
    "Nimari","Ongole","Pulikulam","Rathi","Red_Dane","Red_Sindhi",
    "Sahiwal","Surti","Tharparkar","Toda","Umblachery","Vechur"
]

# ============================================================
# LOAD MODELS
# ============================================================

if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError(f"Model not found: {MODEL_PATH}")

model = tf.keras.models.load_model(MODEL_PATH, compile=False)

if model.output_shape[-1] != len(CLASS_NAMES):
    raise ValueError("Model output and class count do not match!")

# Model for checking whether image is cattle
cow_model = tf.keras.applications.MobileNetV2(weights="imagenet")

# ============================================================
# PREDICTION
# ============================================================

def predict(path):

    img = cv2.imread(path)

    if img is None:
        raise ValueError("Cannot read image")

    # --------------------------------------------------------
    # STEP 1: CHECK WHETHER IMAGE IS CATTLE
    # --------------------------------------------------------

    check = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    check = cv2.resize(check, IMAGE_SIZE).astype(np.float32)
    check = tf.keras.applications.mobilenet_v2.preprocess_input(check)

    check_pred = cow_model.predict(
        np.expand_dims(check, 0), verbose=0
    )

    labels = [
        x[1].lower()
        for x in tf.keras.applications.mobilenet_v2.decode_predictions(
            check_pred, top=5
        )[0]
    ]

    cattle_words = [
        "ox",
        "water_buffalo",
        "buffalo"
    ]

    if not any(word in label for label in labels for word in cattle_words):
        return None

    # --------------------------------------------------------
    # STEP 2: CATTLE BREED PREDICTION
    # --------------------------------------------------------

    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, IMAGE_SIZE).astype(np.float32)

    img = tf.keras.applications.efficientnet_v2.preprocess_input(img)

    pred = model.predict(
        np.expand_dims(img, 0), verbose=0
    )[0]

    ids = np.argsort(pred)[-3:][::-1]

    return [
        (CLASS_NAMES[i], pred[i] * 100)
        for i in ids
    ]

# ============================================================
# SHOW IMAGE
# ============================================================

def show_image(path):

    global photo

    img = Image.open(path).convert("RGB")

    w, h = img.size

    fw = max(100, image_frame.winfo_width() - 40)
    fh = max(100, image_frame.winfo_height() - 40)

    scale = min(fw / w, fh / h)

    img = img.resize(
        (
            max(1, int(w * scale)),
            max(1, int(h * scale))
        ),
        Image.Resampling.LANCZOS
    )

    photo = ImageTk.PhotoImage(img)

    image_label.config(
        image=photo,
        text=""
    )

    image_label.image = photo

# ============================================================
# SELECT IMAGE
# ============================================================

def select_image():

    path = filedialog.askopenfilename(
        filetypes=[
            ("Images", "*.jpg *.jpeg *.png *.bmp *.webp")
        ]
    )

    if not path:
        return

    try:

        show_image(path)

        results = predict(path)

        # ----------------------------------------------------
        # NOT A COW
        # ----------------------------------------------------

        if results is None:

            prediction_label.config(
                text="❌ Image is not a cow"
            )

            confidence_label.config(
                text="Please upload a clear cattle image"
            )

            top3_label.config(
                text=""
            )

            warning_label.config(
                text="⚠ Image rejected"
            )

            status_label.config(
                text="Please select a cattle image"
            )

            return

        # ----------------------------------------------------
        # COW DETECTED
        # ----------------------------------------------------

        prediction_label.config(
            text=f"Predicted Breed: "
                 f"{results[0][0].replace('_',' ')}"
        )

        confidence_label.config(
            text=f"Confidence: {results[0][1]:.2f}%"
        )

        top3_label.config(
            text="\n".join(
                f"{i}. {breed.replace('_',' ')} — {conf:.2f}%"
                for i, (breed, conf) in enumerate(results, 1)
            )
        )

        warning_label.config(
            text=(
                "⚠ Low confidence — try a clearer image."
                if results[0][1] < 50
                else "✓ Prediction completed"
            )
        )

        status_label.config(
            text="Analysis complete"
        )

    except Exception as e:

        messagebox.showerror(
            "Prediction Error",
            str(e)
        )

# ============================================================
# RESET
# ============================================================

def reset():

    image_label.config(
        image="",
        text="No image selected"
    )

    image_label.image = None

    prediction_label.config(
        text="Predicted Breed: ---"
    )

    confidence_label.config(
        text="Confidence: ---"
    )

    top3_label.config(
        text="1. ---\n2. ---\n3. ---"
    )

    warning_label.config(text="")

    status_label.config(
        text="Select an image to begin"
    )

# ============================================================
# GUI
# ============================================================

root = Tk()

root.title(
    "Cattle Breed Identification System"
)

try:
    root.state("zoomed")
except:
    root.geometry("1100x750")

root.minsize(
    900,
    650
)

Label(
    root,
    text="🐄 Cattle Breed Identification",
    font=("Arial", 25, "bold")
).pack(pady=(12, 2))

Label(
    root,
    text="AI-powered identification of 41 cattle breeds",
    font=("Arial", 12)
).pack()

# ------------------------------------------------------------
# IMAGE AREA
# ------------------------------------------------------------

image_frame = Frame(
    root,
    width=750,
    height=400,
    relief="groove",
    bd=2
)

image_frame.pack(pady=8)

image_frame.pack_propagate(False)

image_label = Label(
    image_frame,
    text="No image selected",
    font=("Arial", 16)
)

image_label.pack(
    expand=True,
    fill="both"
)

# ------------------------------------------------------------
# BUTTONS
# ------------------------------------------------------------

buttons = Frame(root)

buttons.pack(pady=8)

Button(
    buttons,
    text="📷 Select Cattle Image",
    font=("Arial", 13, "bold"),
    command=select_image,
    padx=20,
    pady=7
).grid(
    row=0,
    column=0,
    padx=8
)

Button(
    buttons,
    text="🔄 Reset",
    font=("Arial", 13),
    command=reset,
    padx=25,
    pady=7
).grid(
    row=0,
    column=1,
    padx=8
)

# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

results = Frame(root)

results.pack()

prediction_label = Label(
    results,
    text="Predicted Breed: ---",
    font=("Arial", 18, "bold")
)

prediction_label.pack()

confidence_label = Label(
    results,
    text="Confidence: ---",
    font=("Arial", 13)
)

confidence_label.pack()

Label(
    results,
    text="Top 3 Predictions",
    font=("Arial", 12, "bold")
).pack(pady=(5, 0))

top3_label = Label(
    results,
    text="1. ---\n2. ---\n3. ---",
    font=("Arial", 11),
    justify="center"
)

top3_label.pack()

warning_label = Label(
    results,
    font=("Arial", 10)
)

warning_label.pack()

status_label = Label(
    results,
    text="Select an image to begin",
    font=("Arial", 9)
)

status_label.pack()

# ============================================================
# START
# ============================================================

print("System ready.")

root.mainloop()
```

```python
import os
import numpy as np
import tensorflow as tf
import cv2
from tkinter import Tk, Label, Button, filedialog, messagebox, Frame
from PIL import Image, ImageTk

# ============================================================
# SETTINGS
# ============================================================

MODEL_PATH = "Best_Cattle_Breed_Improved.h5"
IMAGE_SIZE = (224, 224)

CLASS_NAMES = [
    "Alambadi", "Amritmahal", "Ayrshire", "Banni", "Bargur", "Bhadawari",
    "Brown_Swiss", "Dangi", "Deoni", "Gir", "Guernsey", "Hallikar", "Hariana",
    "Holstein_Friesian", "Jaffrabadi", "Jersey", "Kangayam", "Kankrej",
    "Kasargod", "Kenkatha", "Kherigarh", "Khillari", "Krishna_Valley",
    "Malnad_gidda", "Mehsana", "Murrah", "Nagori", "Nagpuri", "Nili_Ravi",
    "Nimari", "Ongole", "Pulikulam", "Rathi", "Red_Dane", "Red_Sindhi",
    "Sahiwal", "Surti", "Tharparkar", "Toda", "Umblachery", "Vechur"
]

# ============================================================
# LOAD MODELS
# ============================================================

if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError(
        f"Model not found: {MODEL_PATH}"
    )

model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

if model.output_shape[-1] != len(CLASS_NAMES):
    raise ValueError(
        "Model output and class count do not match!"
    )

# Model used to check whether uploaded image contains cattle
cow_model = tf.keras.applications.MobileNetV2(
    weights="imagenet"
)

# ============================================================
# PREDICTION
# ============================================================

def predict(path):

    img = cv2.imread(path)

    if img is None:
        raise ValueError("Cannot read image")

    # --------------------------------------------------------
    # STEP 1: CHECK WHETHER IMAGE IS CATTLE
    # --------------------------------------------------------

    check = cv2.cvtColor(
        img,
        cv2.COLOR_BGR2RGB
    )

    check = cv2.resize(
        check,
        IMAGE_SIZE
    ).astype(np.float32)

    check = tf.keras.applications.mobilenet_v2.preprocess_input(
        check
    )

    check_pred = cow_model.predict(
        np.expand_dims(check, 0),
        verbose=0
    )

    labels = [
        x[1].lower()
        for x in tf.keras.applications.mobilenet_v2.decode_predictions(
            check_pred,
            top=5
        )[0]
    ]

    cattle_words = [
        "ox",
        "water_buffalo",
        "buffalo"
    ]

    if not any(
        word in label
        for label in labels
        for word in cattle_words
    ):
        return None

    # --------------------------------------------------------
    # STEP 2: CATTLE BREED PREDICTION
    # --------------------------------------------------------

    img = cv2.cvtColor(
        img,
        cv2.COLOR_BGR2RGB
    )

    img = cv2.resize(
        img,
        IMAGE_SIZE
    ).astype(np.float32)

    img = tf.keras.applications.efficientnet_v2.preprocess_input(
        img
    )

    pred = model.predict(
        np.expand_dims(img, 0),
        verbose=0
    )[0]

    # Get top 3 predictions
    ids = np.argsort(pred)[-3:][::-1]

    return [
        (
            CLASS_NAMES[i],
            pred[i] * 100
        )
        for i in ids
    ]


# ============================================================
# SHOW IMAGE
# ============================================================

def show_image(path):

    global photo

    img = Image.open(path).convert("RGB")

    w, h = img.size

    fw = max(
        100,
        image_frame.winfo_width() - 40
    )

    fh = max(
        100,
        image_frame.winfo_height() - 40
    )

    scale = min(
        fw / w,
        fh / h
    )

    img = img.resize(
        (
            max(1, int(w * scale)),
            max(1, int(h * scale))
        ),
        Image.Resampling.LANCZOS
    )

    photo = ImageTk.PhotoImage(img)

    image_label.config(
        image=photo,
        text=""
    )

    image_label.image = photo


# ============================================================
# SELECT IMAGE
# ============================================================

def select_image():

    path = filedialog.askopenfilename(
        filetypes=[
            (
                "Images",
                "*.jpg *.jpeg *.png *.bmp *.webp"
            )
        ]
    )

    if not path:
        return

    try:

        # Show selected image
        show_image(path)

        # Run prediction
        results = predict(path)

        # ----------------------------------------------------
        # IMAGE IS NOT OF A CATTLE BREED
        # ----------------------------------------------------

        if results is None:

            prediction_label.config(
                text="❌ Image is not of a cattle breed"
            )

            confidence_label.config(
                text="Please upload a clear cattle image"
            )

            top3_label.config(
                text=""
            )

            warning_label.config(
                text="⚠ Image rejected"
            )

            status_label.config(
                text="Please select a cattle image"
            )

            return

        # ----------------------------------------------------
        # CATTLE DETECTED
        # ----------------------------------------------------

        prediction_label.config(
            text=(
                f"Predicted Breed: "
                f"{results[0][0].replace('_', ' ')}"
            )
        )

        confidence_label.config(
            text=(
                f"Confidence: "
                f"{results[0][1]:.2f}%"
            )
        )

        # ----------------------------------------------------
        # TOP 3 PREDICTIONS
        # ----------------------------------------------------

        top3_label.config(
            text="\n".join(
                f"{i}. {breed.replace('_', ' ')} — {conf:.2f}%"
                for i, (breed, conf)
                in enumerate(results, 1)
            )
        )

        # ----------------------------------------------------
        # CONFIDENCE WARNING
        # ----------------------------------------------------

        warning_label.config(
            text=(
                "⚠ Low confidence — try a clearer image."
                if results[0][1] < 50
                else "✓ Prediction completed"
            )
        )

        status_label.config(
            text="Analysis complete"
        )

    except Exception as e:

        messagebox.showerror(
            "Prediction Error",
            str(e)
        )


# ============================================================
# RESET
# ============================================================

def reset():

    image_label.config(
        image="",
        text="No image selected"
    )

    image_label.image = None

    prediction_label.config(
        text="Predicted Breed: ---"
    )

    confidence_label.config(
        text="Confidence: ---"
    )

    top3_label.config(
        text="1. ---\n2. ---\n3. ---"
    )

    warning_label.config(
        text=""
    )

    status_label.config(
        text="Select an image to begin"
    )


# ============================================================
# GUI
# ============================================================

root = Tk()

root.title(
    "Cattle Breed Identification System"
)

# Try fullscreen/maximized window
try:
    root.state("zoomed")
except:
    root.geometry("1100x750")

root.minsize(
    900,
    650
)


# ============================================================
# TITLE
# ============================================================

Label(
    root,
    text="🐄 Cattle Breed Identification",
    font=("Arial", 25, "bold")
).pack(
    pady=(12, 2)
)

Label(
    root,
    text="AI-powered identification of 41 cattle breeds",
    font=("Arial", 12)
).pack()


# ============================================================
# IMAGE AREA
# ============================================================

image_frame = Frame(
    root,
    width=750,
    height=400,
    relief="groove",
    bd=2
)

image_frame.pack(
    pady=8
)

image_frame.pack_propagate(False)

image_label = Label(
    image_frame,
    text="No image selected",
    font=("Arial", 16)
)

image_label.pack(
    expand=True,
    fill="both"
)


# ============================================================
# BUTTONS
# ============================================================

buttons = Frame(root)

buttons.pack(
    pady=8
)

Button(
    buttons,
    text="📷 Select Cattle Image",
    font=("Arial", 13, "bold"),
    command=select_image,
    padx=20,
    pady=7
).grid(
    row=0,
    column=0,
    padx=8
)

Button(
    buttons,
    text="🔄 Reset",
    font=("Arial", 13),
    command=reset,
    padx=25,
    pady=7
).grid(
    row=0,
    column=1,
    padx=8
)


# ============================================================
# RESULTS
# ============================================================

results = Frame(root)

results.pack()


# Predicted breed
prediction_label = Label(
    results,
    text="Predicted Breed: ---",
    font=("Arial", 18, "bold")
)

prediction_label.pack()


# Confidence
confidence_label = Label(
    results,
    text="Confidence: ---",
    font=("Arial", 13)
)

confidence_label.pack()


# Top 3 heading
Label(
    results,
    text="Top 3 Predictions",
    font=("Arial", 12, "bold")
).pack(
    pady=(5, 0)
)


# Top 3 predictions
top3_label = Label(
    results,
    text="1. ---\n2. ---\n3. ---",
    font=("Arial", 11),
    justify="center"
)

top3_label.pack()


# Warning / status
warning_label = Label(
    results,
    font=("Arial", 10)
)

warning_label.pack()


# System status
status_label = Label(
    results,
    text="Select an image to begin",
    font=("Arial", 9)
)

status_label.pack()


# ============================================================
# START SYSTEM
# ============================================================

print("System ready.")

root.mainloop()
```
