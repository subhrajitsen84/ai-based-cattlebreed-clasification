```python
import os
import numpy as np
import cv2
import tensorflow as tf
import warnings

from tkinter import Tk, Label, Button, filedialog, messagebox
from PIL import Image, ImageTk

warnings.filterwarnings("ignore")


# ============================================================
# CONSTANTS
# ============================================================

MODEL_PATH = "Best_Cattle_Breed (1).h5"
DATA_DIR = "data"


# ============================================================
# CHECK FILES
# ============================================================

if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError(
        f"Model file not found:\n{MODEL_PATH}\n\n"
        "Make sure the model is in the same folder as this notebook."
    )

if not os.path.exists(DATA_DIR):
    raise FileNotFoundError(
        f"Data folder not found:\n{DATA_DIR}\n\n"
        "Make sure the data folder is in the same folder as this notebook."
    )


# ============================================================
# LOAD MODEL
# ============================================================

print("Loading cattle breed model...")

model = tf.keras.models.load_model(MODEL_PATH)

print("Model loaded successfully!")


# ============================================================
# LOAD CLASS NAMES
# ============================================================

CLASS_NAMES = sorted(
    [
        folder
        for folder in os.listdir(DATA_DIR)
        if os.path.isdir(os.path.join(DATA_DIR, folder))
    ]
)

if len(CLASS_NAMES) == 0:
    raise ValueError(
        "No breed folders were found inside the data folder."
    )

print("\nCattle breeds found:")

for i, class_name in enumerate(CLASS_NAMES):
    print(f"{i}: {class_name}")

print(f"\nTotal classes: {len(CLASS_NAMES)}")


# ============================================================
# IMAGE PREPROCESSING
# ============================================================

def preprocess_image(image_path):

    img = cv2.imread(image_path)

    if img is None:
        raise ValueError("Could not read the selected image.")

    # Convert BGR to RGB
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Resize to model input size
    img = cv2.resize(img, (224, 224))

    # Convert to float32
    img = img.astype(np.float32)

    # EfficientNetV2 preprocessing
    img = tf.keras.applications.efficientnet_v2.preprocess_input(img)

    # Add batch dimension
    img = np.expand_dims(img, axis=0)

    return img


# ============================================================
# PREDICTION
# ============================================================

def predict_image(image_path):

    try:

        img = preprocess_image(image_path)

        predictions = model.predict(img, verbose=0)

        probabilities = predictions[0]

        class_id = int(np.argmax(probabilities))

        if class_id >= len(CLASS_NAMES):
            raise ValueError(
                "The number of model output classes does not match "
                "the number of breed folders in the data folder."
            )

        predicted_class = CLASS_NAMES[class_id]

        confidence = float(probabilities[class_id]) * 100

        return predicted_class, confidence

    except Exception as e:

        messagebox.showerror(
            "Prediction Error",
            str(e)
        )

        return None, None


# ============================================================
# SELECT IMAGE
# ============================================================

def browse_image():

    file_path = filedialog.askopenfilename(
        title="Select Cattle Image",
        filetypes=[
            ("Image files", "*.jpg *.jpeg *.png *.bmp"),
            ("All files", "*.*")
        ]
    )

    if not file_path:
        return

    try:

        # Open image
        image = Image.open(file_path)

        # Resize image for GUI
        image = image.resize((300, 300))

        # Convert for Tkinter
        photo = ImageTk.PhotoImage(image)

        # Display image
        image_label.config(image=photo)
        image_label.image = photo

        # Predict breed
        predicted_class, confidence = predict_image(file_path)

        if predicted_class is not None:

            result_label.config(
                text=(
                    f"Prediction: {predicted_class}\n"
                    f"Confidence: {confidence:.2f}%"
                )
            )

    except Exception as e:

        messagebox.showerror(
            "Error",
            str(e)
        )


# ============================================================
# GUI
# ============================================================

root = Tk()

root.title("Cattle Breed Classifier")

root.geometry("450x600")

root.resizable(False, False)


# ============================================================
# TITLE
# ============================================================

title_label = Label(
    root,
    text="Indian Cattle Breed Classifier",
    font=("Arial", 18, "bold")
)

title_label.pack(pady=15)


# ============================================================
# IMAGE AREA
# ============================================================

image_label = Label(
    root,
    text="Select a cattle image",
    width=35,
    height=15
)

image_label.pack(pady=10)


# ============================================================
# SELECT BUTTON
# ============================================================

browse_btn = Button(
    root,
    text="Select Cattle Image",
    font=("Arial", 12),
    command=browse_image
)

browse_btn.pack(pady=20)


# ============================================================
# RESULT
# ============================================================

result_label = Label(
    root,
    text="",
    font=("Arial", 14)
)

result_label.pack(pady=15)


# ============================================================
# START PROGRAM
# ============================================================

root.mainloop()
```

```python
import os
import numpy as np
import cv2
import tensorflow as tf
from scipy.optimize import linear_sum_assignment

# ============================================================
# SETTINGS
# ============================================================

DATA_DIR = "data"
IMAGES_PER_BREED = 10

# ============================================================
# GET BREED FOLDERS
# ============================================================

breed_folders = sorted([
    f for f in os.listdir(DATA_DIR)
    if os.path.isdir(os.path.join(DATA_DIR, f))
])

print("Breeds found:", len(breed_folders))
print("Testing up to", IMAGES_PER_BREED, "images per breed...")
print("=" * 60)


# ============================================================
# CREATE SCORE MATRIX
# ============================================================

# Rows = actual breeds
# Columns = model output classes

score_matrix = np.zeros(
    (len(breed_folders), 41),
    dtype=np.float64
)


# ============================================================
# PROCESS EACH BREED
# ============================================================

for breed_index, breed in enumerate(breed_folders):

    folder = os.path.join(DATA_DIR, breed)

    image_files = [
        f for f in os.listdir(folder)
        if f.lower().endswith(
            (".jpg", ".jpeg", ".png", ".bmp")
        )
    ]

    # Use up to IMAGES_PER_BREED images
    image_files = image_files[:IMAGES_PER_BREED]

    valid_images = 0

    for filename in image_files:

        image_path = os.path.join(folder, filename)

        img = cv2.imread(image_path)

        if img is None:
            continue

        # BGR -> RGB
        img = cv2.cvtColor(
            img,
            cv2.COLOR_BGR2RGB
        )

        # Resize
        img = cv2.resize(
            img,
            (224, 224)
        )

        # Float32
        img = img.astype(np.float32)

        # EfficientNetV2 preprocessing
        img = tf.keras.applications.efficientnet_v2.preprocess_input(img)

        # Batch dimension
        img = np.expand_dims(img, axis=0)

        # Prediction
        probabilities = model.predict(
            img,
            verbose=0
        )[0]

        score_matrix[breed_index] += probabilities

        valid_images += 1

    if valid_images > 0:
        score_matrix[breed_index] /= valid_images

    print(
        f"{breed_index:2d}  {breed:20s} "
        f"({valid_images} images tested)"
    )


# ============================================================
# FIND BEST ONE-TO-ONE CLASS MAPPING
# ============================================================

# Hungarian algorithm finds the best assignment between:
# actual breed folders <-> model output indices

row_indices, col_indices = linear_sum_assignment(
    -score_matrix
)


# ============================================================
# DISPLAY MAPPING
# ============================================================

print("\n")
print("=" * 70)
print("RECOVERED MODEL CLASS MAPPING")
print("=" * 70)

class_mapping = {}

for row, col in zip(row_indices, col_indices):

    breed = breed_folders[row]

    confidence = score_matrix[row, col] * 100

    class_mapping[int(col)] = breed

    print(
        f"Model class {col:2d}  ->  "
        f"{breed:20s}  "
        f"average score: {confidence:.2f}%"
    )


# ============================================================
# SAVE MAPPING
# ============================================================

print("\n")
print("=" * 70)
print("CLASS_NAMES FOR YOUR FINAL PROGRAM")
print("=" * 70)

CLASS_NAMES = [
    class_mapping[i]
    for i in range(41)
]

for i, name in enumerate(CLASS_NAMES):
    print(f'"{name}",')


print("\nMapping generated successfully.")
```

```python
import os
import cv2
import numpy as np
import tensorflow as tf

MODEL_PATH = "Best_Cattle_Breed (1).h5"
DATA_DIR = "data"

model = tf.keras.models.load_model(MODEL_PATH)

CLASS_NAMES = sorted([
    x for x in os.listdir(DATA_DIR)
    if os.path.isdir(os.path.join(DATA_DIR, x))
])

correct = 0
total = 0

print("🐄 TESTING CATTLE BREED MODEL")
print("=" * 50)

for breed in CLASS_NAMES[:10]:   # test first 10 breeds

    breed_folder = os.path.join(DATA_DIR, breed)

    images = [
        f for f in os.listdir(breed_folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png", ".webp"))
    ]

    if not images:
        continue

    image_path = os.path.join(breed_folder, images[0])

    img = cv2.imread(image_path)

    if img is None:
        continue

    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (224, 224))

    img_array = np.expand_dims(img, axis=0)

    img_array = tf.keras.applications.efficientnet_v2.preprocess_input(
        img_array
    )

    prediction = model.predict(img_array, verbose=0)[0]

    predicted_index = np.argmax(prediction)
    predicted_breed = CLASS_NAMES[predicted_index]
    confidence = prediction[predicted_index] * 100

    total += 1

    if predicted_breed == breed:
        correct += 1
        result = "✅"
    else:
        result = "❌"

    print(
        f"{result} Actual: {breed:<20} "
        f"Predicted: {predicted_breed:<20} "
        f"Confidence: {confidence:.2f}%"
    )

print("=" * 50)
print(f"Correct: {correct}/{total}")

if total > 0:
    print(f"Accuracy on this small test: {(correct/total)*100:.2f}%")
```

```python
import os
import numpy as np
import tensorflow as tf
import warnings

from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, confusion_matrix

warnings.filterwarnings("ignore")

# ============================================================
# SETTINGS
# ============================================================

DATA_DIR = "data"

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 123

INITIAL_EPOCHS = 15
FINE_TUNE_EPOCHS = 25

INITIAL_LR = 0.001
FINE_TUNE_LR = 0.00001

MODEL_PATH = "Best_Cattle_Breed_Improved.h5"

# ============================================================
# LOAD DATA
# ============================================================

print("Loading training dataset...")

train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR,
    validation_split=0.2,
    subset="training",
    seed=SEED,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=True
)

print("\nLoading validation dataset...")

val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR,
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

CLASS_NAMES = train_ds.class_names
NUM_CLASSES = len(CLASS_NAMES)

print("\n========================================")
print("DATASET INFORMATION")
print("========================================")

print("Number of classes:", NUM_CLASSES)

for i, name in enumerate(CLASS_NAMES):
    print(i, "->", name)

# ============================================================
# PERFORMANCE
# ============================================================

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)

# ============================================================
# CALCULATE CLASS WEIGHTS
# ============================================================

print("\nCalculating class weights...")

class_counts = np.zeros(NUM_CLASSES, dtype=np.int32)

for _, labels in train_ds:
    for label in labels.numpy():
        class_counts[label] += 1

class_weights_array = compute_class_weight(
    class_weight="balanced",
    classes=np.arange(NUM_CLASSES),
    y=np.repeat(
        np.arange(NUM_CLASSES),
        class_counts
    )
)

class_weights = {
    i: float(class_weights_array[i])
    for i in range(NUM_CLASSES)
}

print("\nClass weights:")

for i, name in enumerate(CLASS_NAMES):
    print(
        f"{name:20} : "
        f"{class_counts[i]:4} images | "
        f"weight = {class_weights[i]:.3f}"
    )

# ============================================================
# DATA AUGMENTATION
# ============================================================

data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.15),
    tf.keras.layers.RandomZoom(0.15),
    tf.keras.layers.RandomContrast(0.10),
])

# ============================================================
# BUILD EFFICIENTNETV2B0
# ============================================================

print("\nBuilding EfficientNetV2B0...")

base_model = tf.keras.applications.EfficientNetV2B0(
    input_shape=IMAGE_SIZE + (3,),
    include_top=False,
    weights="imagenet"
)

# Freeze base initially
base_model.trainable = False

inputs = tf.keras.Input(
    shape=IMAGE_SIZE + (3,)
)

x = data_augmentation(inputs)

x = tf.keras.applications.efficientnet_v2.preprocess_input(x)

x = base_model(
    x,
    training=False
)

x = tf.keras.layers.GlobalAveragePooling2D()(x)

x = tf.keras.layers.BatchNormalization()(x)

x = tf.keras.layers.Dropout(0.4)(x)

outputs = tf.keras.layers.Dense(
    NUM_CLASSES,
    activation="softmax"
)(x)

model = tf.keras.Model(
    inputs,
    outputs
)

# ============================================================
# COMPILE - STAGE 1
# ============================================================

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=INITIAL_LR
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("\n========================================")
print("STAGE 1: TRAINING CLASSIFICATION HEAD")
print("========================================")

model.summary()

# ============================================================
# CALLBACKS
# ============================================================

callbacks = [

    tf.keras.callbacks.ModelCheckpoint(
        MODEL_PATH,
        monitor="val_accuracy",
        save_best_only=True,
        verbose=1
    ),

    tf.keras.callbacks.EarlyStopping(
        monitor="val_accuracy",
        patience=5,
        restore_best_weights=True,
        verbose=1
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=2,
        min_lr=1e-7,
        verbose=1
    )
]

# ============================================================
# STAGE 1 TRAINING
# ============================================================

history1 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=INITIAL_EPOCHS,
    class_weight=class_weights,
    callbacks=callbacks
)

# ============================================================
# STAGE 2 - FINE TUNING
# ============================================================

print("\n========================================")
print("STAGE 2: FINE-TUNING EFFICIENTNET")
print("========================================")

base_model.trainable = True

# Freeze the first ~70% of layers
fine_tune_from = int(len(base_model.layers) * 0.70)

for layer in base_model.layers[:fine_tune_from]:
    layer.trainable = False

for layer in base_model.layers[fine_tune_from:]:
    layer.trainable = True

print(
    "Trainable layers:",
    sum(
        layer.trainable
        for layer in base_model.layers
    )
)

# Recompile with very small learning rate

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=FINE_TUNE_LR
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# ============================================================
# FINE-TUNING
# ============================================================

history2 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=FINE_TUNE_EPOCHS,
    class_weight=class_weights,
    callbacks=callbacks
)

# ============================================================
# LOAD BEST MODEL
# ============================================================

print("\nLoading best model...")

model = tf.keras.models.load_model(
    MODEL_PATH
)

print("✅ Best model loaded")

# ============================================================
# FINAL EVALUATION
# ============================================================

print("\n========================================")
print("FINAL MODEL EVALUATION")
print("========================================")

loss, accuracy = model.evaluate(
    val_ds,
    verbose=1
)

print(
    f"\n✅ Final Validation Accuracy: "
    f"{accuracy * 100:.2f}%"
)

# ============================================================
# DETAILED PREDICTIONS
# ============================================================

print("\nCalculating detailed predictions...")

y_true = []
y_pred = []

for images, labels in val_ds:

    predictions = model.predict(
        images,
        verbose=0
    )

    y_true.extend(
        labels.numpy()
    )

    y_pred.extend(
        np.argmax(
            predictions,
            axis=1
        )
    )

y_true = np.array(y_true)
y_pred = np.array(y_pred)

# ============================================================
# CLASSIFICATION REPORT
# ============================================================

print("\n========================================")
print("CLASSIFICATION REPORT")
print("========================================\n")

print(
    classification_report(
        y_true,
        y_pred,
        labels=np.arange(NUM_CLASSES),
        target_names=CLASS_NAMES,
        digits=3,
        zero_division=0
    )
)

# ============================================================
# CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=np.arange(NUM_CLASSES)
)

print("\n========================================")
print("TOP 20 WRONG PREDICTIONS")
print("========================================\n")

confusions = []

for i in range(NUM_CLASSES):

    for j in range(NUM_CLASSES):

        if i != j and cm[i, j] > 0:

            confusions.append(
                (
                    cm[i, j],
                    CLASS_NAMES[i],
                    CLASS_NAMES[j]
                )
            )

confusions.sort(
    reverse=True
)

for count, actual, predicted in confusions[:20]:

    print(
        f"{actual:20} → "
        f"{predicted:20} : "
        f"{count} images"
    )

print("\n========================================")
print("TRAINING COMPLETE")
print("========================================")

print(
    f"Best model saved as: {MODEL_PATH}"
)

print(
    f"Final validation accuracy: "
    f"{accuracy * 100:.2f}%"
)
```

```python
import os
import numpy as np
import tensorflow as tf
import cv2
import warnings

from tkinter import (
    Tk, Label, Button, filedialog,
    messagebox, Frame
)

from PIL import Image, ImageTk

warnings.filterwarnings("ignore")


# ============================================================
# SETTINGS
# ============================================================

MODEL_PATH = "Best_Cattle_Breed_Improved.h5"

IMAGE_SIZE = (224, 224)

# Confidence below this value will show a warning
CONFIDENCE_THRESHOLD = 50.0


# ============================================================
# 41 CATTLE BREEDS
# ============================================================

CLASS_NAMES = [
    "Alambadi",
    "Amritmahal",
    "Ayrshire",
    "Banni",
    "Bargur",
    "Bhadawari",
    "Brown_Swiss",
    "Dangi",
    "Deoni",
    "Gir",
    "Guernsey",
    "Hallikar",
    "Hariana",
    "Holstein_Friesian",
    "Jaffrabadi",
    "Jersey",
    "Kangayam",
    "Kankrej",
    "Kasargod",
    "Kenkatha",
    "Kherigarh",
    "Khillari",
    "Krishna_Valley",
    "Malnad_gidda",
    "Mehsana",
    "Murrah",
    "Nagori",
    "Nagpuri",
    "Nili_Ravi",
    "Nimari",
    "Ongole",
    "Pulikulam",
    "Rathi",
    "Red_Dane",
    "Red_Sindhi",
    "Sahiwal",
    "Surti",
    "Tharparkar",
    "Toda",
    "Umblachery",
    "Vechur"
]


# ============================================================
# CHECK MODEL
# ============================================================

print("Checking model...")

if not os.path.exists(MODEL_PATH):

    print("❌ Model not found:")
    print(os.path.abspath(MODEL_PATH))

    raise FileNotFoundError(
        f"{MODEL_PATH} was not found."
    )

print("✅ Model found")
print("Loading model...")


# ============================================================
# LOAD MODEL
# ============================================================

try:

    model = tf.keras.models.load_model(
        MODEL_PATH,
        compile=False
    )

    print("✅ Model loaded successfully")

except Exception as e:

    print("❌ Could not load model")
    print(e)

    raise


# ============================================================
# VERIFY OUTPUT
# ============================================================

try:

    output_classes = model.output_shape[-1]

    print(
        f"Model output classes: {output_classes}"
    )

    if output_classes != len(CLASS_NAMES):

        raise ValueError(
            f"Model has {output_classes} outputs "
            f"but CLASS_NAMES contains "
            f"{len(CLASS_NAMES)} classes."
        )

except Exception as e:

    print("❌ Model/class mismatch")
    print(e)

    raise


# ============================================================
# IMAGE PREPROCESSING
# ============================================================

def preprocess_image(image_path):

    image = cv2.imread(image_path)

    if image is None:

        raise ValueError(
            "Could not read the selected image."
        )

    # BGR → RGB
    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    # Resize to model input
    image = cv2.resize(
        image,
        IMAGE_SIZE
    )

    image = image.astype(
        np.float32
    )

    # Same preprocessing used during training
    image = tf.keras.applications.efficientnet_v2.preprocess_input(
        image
    )

    # Add batch dimension
    image = np.expand_dims(
        image,
        axis=0
    )

    return image


# ============================================================
# PREDICTION
# ============================================================

def predict_breed(image_path):

    processed_image = preprocess_image(
        image_path
    )

    predictions = model.predict(
        processed_image,
        verbose=0
    )[0]

    # Get top 3 predictions
    top_indices = np.argsort(
        predictions
    )[-3:][::-1]

    results = []

    for index in top_indices:

        breed = CLASS_NAMES[index]

        confidence = (
            float(predictions[index]) * 100
        )

        results.append(
            (breed, confidence)
        )

    return results


# ============================================================
# GUI FUNCTIONS
# ============================================================

def select_image():

    file_path = filedialog.askopenfilename(

        title="Select Cattle Image",

        filetypes=[
            (
                "Image files",
                "*.jpg *.jpeg *.png *.bmp *.webp"
            ),
            (
                "All files",
                "*.*"
            )
        ]
    )

    if not file_path:
        return

    try:

        # ----------------------------------------------------
        # SHOW IMAGE
        # ----------------------------------------------------

        image = Image.open(
            file_path
        )

        image.thumbnail(
            (500, 400)
        )

        photo = ImageTk.PhotoImage(
            image
        )

        image_label.config(
            image=photo
        )

        image_label.image = photo

        # ----------------------------------------------------
        # RUN PREDICTION
        # ----------------------------------------------------

        status_label.config(
            text="Analyzing image..."
        )

        root.update()

        results = predict_breed(
            file_path
        )

        # ----------------------------------------------------
        # MAIN PREDICTION
        # ----------------------------------------------------

        best_breed = results[0][0]

        best_confidence = results[0][1]

        prediction_label.config(
            text=f"Predicted Breed:\n{best_breed}"
        )

        confidence_label.config(
            text=f"Confidence: {best_confidence:.2f}%"
        )

        # ----------------------------------------------------
        # TOP 3
        # ----------------------------------------------------

        top3_text = "Top 3 Predictions\n\n"

        for i, (breed, confidence) in enumerate(
            results,
            start=1
        ):

            top3_text += (
                f"{i}. {breed.replace('_', ' ')}"
                f"  —  {confidence:.2f}%\n"
            )

        top3_label.config(
            text=top3_text
        )

        # ----------------------------------------------------
        # CONFIDENCE WARNING
        # ----------------------------------------------------

        if best_confidence < CONFIDENCE_THRESHOLD:

            warning_label.config(
                text=(
                    "⚠ Low confidence\n"
                    "Try a clearer cattle image."
                )
            )

        else:

            warning_label.config(
                text="✓ Prediction completed"
            )

        status_label.config(
            text="Analysis complete"
        )

    except Exception as e:

        messagebox.showerror(
            "Prediction Error",
            str(e)
        )

        status_label.config(
            text="Error while analyzing image"
        )


# ============================================================
# RESET
# ============================================================

def reset():

    image_label.config(
        image=""
    )

    image_label.image = None

    prediction_label.config(
        text="Predicted Breed:\n---"
    )

    confidence_label.config(
        text="Confidence: ---"
    )

    top3_label.config(
        text="Top 3 Predictions\n\n---"
    )

    warning_label.config(
        text=""
    )

    status_label.config(
        text="Select an image to begin"
    )


# ============================================================
# MAIN WINDOW
# ============================================================

root = Tk()

root.title(
    "Cattle Breed Identification System"
)

root.geometry(
    "900x750"
)

root.resizable(
    False,
    False
)


# ============================================================
# TITLE
# ============================================================

title_label = Label(
    root,
    text="🐄 Cattle Breed Identification",
    font=("Arial", 24, "bold")
)

title_label.pack(
    pady=15
)


subtitle_label = Label(
    root,
    text=(
        "AI-powered identification of 41 cattle breeds"
    ),
    font=("Arial", 12)
)

subtitle_label.pack(
    pady=5
)


# ============================================================
# IMAGE AREA
# ============================================================

image_frame = Frame(
    root
)

image_frame.pack(
    pady=15
)


image_label = Label(
    image_frame,
    text="No image selected",
    width=55,
    height=18,
    relief="groove"
)

image_label.pack()


# ============================================================
# BUTTONS
# ============================================================

button_frame = Frame(
    root
)

button_frame.pack(
    pady=10
)


select_button = Button(
    button_frame,
    text="📷 Select Cattle Image",
    font=("Arial", 13, "bold"),
    command=select_image,
    padx=20,
    pady=8
)

select_button.grid(
    row=0,
    column=0,
    padx=10
)


reset_button = Button(
    button_frame,
    text="🔄 Reset",
    font=("Arial", 13),
    command=reset,
    padx=25,
    pady=8
)

reset_button.grid(
    row=0,
    column=1,
    padx=10
)


# ============================================================
# RESULT AREA
# ============================================================

prediction_label = Label(
    root,
    text="Predicted Breed:\n---",
    font=("Arial", 19, "bold")
)

prediction_label.pack(
    pady=8
)


confidence_label = Label(
    root,
    text="Confidence: ---",
    font=("Arial", 14)
)

confidence_label.pack(
    pady=3
)


top3_label = Label(
    root,
    text="Top 3 Predictions\n\n---",
    font=("Arial", 12),
    justify="left"
)

top3_label.pack(
    pady=8
)


warning_label = Label(
    root,
    text="",
    font=("Arial", 11)
)

warning_label.pack(
    pady=5
)


# ============================================================
# STATUS
# ============================================================

status_label = Label(
    root,
    text="Select an image to begin",
    font=("Arial", 10)
)

status_label.pack(
    pady=10
)


# ============================================================
# START APPLICATION
# ============================================================

print("\n================================")
print("CATTLE BREED CLASSIFIER")
print("================================")

print(
    f"Model: {MODEL_PATH}"
)

print(
    f"Classes: {len(CLASS_NAMES)}"
)

print("Ready!")

root.mainloop()
```

```python
import os
import numpy as np
import tensorflow as tf
import cv2
import warnings

from tkinter import (
    Tk,
    Label,
    Button,
    filedialog,
    messagebox,
    Frame,
    Canvas
)

from PIL import Image, ImageTk

warnings.filterwarnings("ignore")


# ============================================================
# SETTINGS
# ============================================================

MODEL_PATH = "Best_Cattle_Breed_Improved.h5"

IMAGE_SIZE = (224, 224)

# Below this confidence, show a warning
CONFIDENCE_THRESHOLD = 50.0


# ============================================================
# 41 CATTLE BREEDS
# ============================================================

CLASS_NAMES = [
    "Alambadi",
    "Amritmahal",
    "Ayrshire",
    "Banni",
    "Bargur",
    "Bhadawari",
    "Brown_Swiss",
    "Dangi",
    "Deoni",
    "Gir",
    "Guernsey",
    "Hallikar",
    "Hariana",
    "Holstein_Friesian",
    "Jaffrabadi",
    "Jersey",
    "Kangayam",
    "Kankrej",
    "Kasargod",
    "Kenkatha",
    "Kherigarh",
    "Khillari",
    "Krishna_Valley",
    "Malnad_gidda",
    "Mehsana",
    "Murrah",
    "Nagori",
    "Nagpuri",
    "Nili_Ravi",
    "Nimari",
    "Ongole",
    "Pulikulam",
    "Rathi",
    "Red_Dane",
    "Red_Sindhi",
    "Sahiwal",
    "Surti",
    "Tharparkar",
    "Toda",
    "Umblachery",
    "Vechur"
]


# ============================================================
# CHECK MODEL
# ============================================================

print("========================================")
print("CATTLE BREED IDENTIFICATION SYSTEM")
print("========================================")

print("\nChecking model...")

if not os.path.exists(MODEL_PATH):

    print("❌ MODEL NOT FOUND")
    print()
    print("Expected model:")
    print(os.path.abspath(MODEL_PATH))
    print()

    raise FileNotFoundError(
        "Best_Cattle_Breed_Improved.h5 was not found."
    )

print("✅ Model found")
print("Loading model...")


# ============================================================
# LOAD MODEL
# ============================================================

try:

    model = tf.keras.models.load_model(
        MODEL_PATH,
        compile=False
    )

    print("✅ Model loaded successfully")

except Exception as e:

    print("❌ Error loading model:")
    print(e)

    raise


# ============================================================
# CHECK MODEL OUTPUT
# ============================================================

model_output_classes = model.output_shape[-1]

print(
    f"Model output classes: {model_output_classes}"
)

print(
    f"Expected classes: {len(CLASS_NAMES)}"
)

if model_output_classes != len(CLASS_NAMES):

    raise ValueError(
        f"\nModel/class mismatch!\n"
        f"Model outputs: {model_output_classes}\n"
        f"Class names: {len(CLASS_NAMES)}"
    )

print("✅ Model has 41 outputs")
print("✅ Class list matches model")


# ============================================================
# IMAGE PREPROCESSING
# ============================================================

def preprocess_image(image_path):

    image = cv2.imread(image_path)

    if image is None:

        raise ValueError(
            "Unable to read the selected image."
        )

    # OpenCV uses BGR
    # Convert BGR -> RGB

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    # Resize for the model

    image = cv2.resize(
        image,
        IMAGE_SIZE
    )

    # Convert to float

    image = image.astype(
        np.float32
    )

    # Same preprocessing used during training

    image = (
        tf.keras.applications.efficientnet_v2
        .preprocess_input(image)
    )

    # Add batch dimension

    image = np.expand_dims(
        image,
        axis=0
    )

    return image


# ============================================================
# PREDICT BREED
# ============================================================

def predict_breed(image_path):

    image = preprocess_image(
        image_path
    )

    predictions = model.predict(
        image,
        verbose=0
    )[0]

    # Get indexes of highest 3 probabilities

    top_indices = np.argsort(
        predictions
    )[-3:][::-1]

    results = []

    for index in top_indices:

        breed = CLASS_NAMES[index]

        confidence = (
            float(predictions[index]) * 100
        )

        results.append(
            (
                breed,
                confidence
            )
        )

    return results


# ============================================================
# DISPLAY IMAGE
# ============================================================

def display_image(image_path):

    image = Image.open(
        image_path
    )

    # Convert image to RGB/RGBA

    if image.mode not in (
        "RGB",
        "RGBA"
    ):

        image = image.convert("RGB")

    # Maximum display size

    MAX_WIDTH = 520
    MAX_HEIGHT = 350

    original_width, original_height = image.size

    # Calculate scale while maintaining aspect ratio

    scale = min(
        MAX_WIDTH / original_width,
        MAX_HEIGHT / original_height
    )

    new_width = int(
        original_width * scale
    )

    new_height = int(
        original_height * scale
    )

    # Make sure dimensions aren't zero

    new_width = max(
        1,
        new_width
    )

    new_height = max(
        1,
        new_height
    )

    # Resize image

    image = image.resize(
        (new_width, new_height),
        Image.Resampling.LANCZOS
    )

    photo = ImageTk.PhotoImage(
        image
    )

    # Display image

    image_label.config(
        image=photo,
        text=""
    )

    # IMPORTANT:
    # Keep reference to prevent Tkinter
    # from deleting the image.

    image_label.image = photo


# ============================================================
# SELECT IMAGE
# ============================================================

def select_image():

    file_path = filedialog.askopenfilename(

        title="Select Cattle Image",

        filetypes=[
            (
                "Image Files",
                "*.jpg *.jpeg *.png *.bmp *.webp"
            ),
            (
                "All Files",
                "*.*"
            )
        ]
    )

    if not file_path:

        return

    try:

        # --------------------------------------------
        # SHOW SELECTED IMAGE
        # --------------------------------------------

        display_image(
            file_path
        )

        # --------------------------------------------
        # UPDATE STATUS
        # --------------------------------------------

        status_label.config(
            text="Analyzing image..."
        )

        root.update_idletasks()

        # --------------------------------------------
        # PREDICT
        # --------------------------------------------

        results = predict_breed(
            file_path
        )

        # --------------------------------------------
        # BEST PREDICTION
        # --------------------------------------------

        best_breed = results[0][0]

        best_confidence = results[0][1]

        prediction_label.config(
            text=(
                "Predicted Breed:\n"
                + best_breed.replace(
                    "_",
                    " "
                )
            )
        )

        confidence_label.config(
            text=(
                f"Confidence: "
                f"{best_confidence:.2f}%"
            )
        )

        # --------------------------------------------
        # TOP 3 PREDICTIONS
        # --------------------------------------------

        top3_text = (
            "Top 3 Predictions\n\n"
        )

        for i, (
            breed,
            confidence
        ) in enumerate(
            results,
            start=1
        ):

            clean_breed = breed.replace(
                "_",
                " "
            )

            top3_text += (
                f"{i}. {clean_breed}"
                f"  —  {confidence:.2f}%\n"
            )

        top3_label.config(
            text=top3_text
        )

        # --------------------------------------------
        # CONFIDENCE WARNING
        # --------------------------------------------

        if best_confidence < CONFIDENCE_THRESHOLD:

            warning_label.config(
                text=(
                    "⚠ LOW CONFIDENCE\n"
                    "Try a clearer cattle image."
                )
            )

        else:

            warning_label.config(
                text=(
                    "✓ Prediction completed"
                )
            )

        # --------------------------------------------
        # STATUS
        # --------------------------------------------

        status_label.config(
            text="Analysis complete"
        )

    except Exception as e:

        messagebox.showerror(
            "Prediction Error",
            str(e)
        )

        status_label.config(
            text="Prediction failed"
        )


# ============================================================
# RESET
# ============================================================

def reset():

    image_label.config(
        image="",
        text="No image selected"
    )

    image_label.image = None

    prediction_label.config(
        text="Predicted Breed:\n---"
    )

    confidence_label.config(
        text="Confidence: ---"
    )

    top3_label.config(
        text="Top 3 Predictions\n\n---"
    )

    warning_label.config(
        text=""
    )

    status_label.config(
        text="Select an image to begin"
    )


# ============================================================
# CREATE WINDOW
# ============================================================

root = Tk()

root.title(
    "Cattle Breed Identification System"
)

root.geometry(
    "900x800"
)

root.resizable(
    False,
    False
)


# ============================================================
# TITLE
# ============================================================

title_label = Label(
    root,
    text="🐄 Cattle Breed Identification",
    font=(
        "Arial",
        24,
        "bold"
    )
)

title_label.pack(
    pady=(18, 5)
)


subtitle_label = Label(
    root,
    text=(
        "AI-powered identification of 41 cattle breeds"
    ),
    font=(
        "Arial",
        12
    )
)

subtitle_label.pack(
    pady=(0, 12)
)


# ============================================================
# IMAGE FRAME
# ============================================================

image_frame = Frame(
    root,
    width=560,
    height=380,
    relief="groove",
    borderwidth=2
)

image_frame.pack(
    pady=5
)

# Prevent frame from shrinking to image size

image_frame.pack_propagate(
    False
)


# ============================================================
# IMAGE LABEL
# ============================================================

image_label = Label(
    image_frame,
    text="No image selected",
    font=(
        "Arial",
        13
    ),
    anchor="center"
)

image_label.pack(
    expand=True,
    fill="both"
)


# ============================================================
# BUTTON FRAME
# ============================================================

button_frame = Frame(
    root
)

button_frame.pack(
    pady=12
)


# ============================================================
# SELECT BUTTON
# ============================================================

select_button = Button(
    button_frame,
    text="📷 Select Cattle Image",
    font=(
        "Arial",
        13,
        "bold"
    ),
    command=select_image,
    padx=20,
    pady=8
)

select_button.grid(
    row=0,
    column=0,
    padx=10
)


# ============================================================
# RESET BUTTON
# ============================================================

reset_button = Button(
    button_frame,
    text="🔄 Reset",
    font=(
        "Arial",
        13
    ),
    command=reset,
    padx=25,
    pady=8
)

reset_button.grid(
    row=0,
    column=1,
    padx=10
)


# ============================================================
# PREDICTION
# ============================================================

prediction_label = Label(
    root,
    text="Predicted Breed:\n---",
    font=(
        "Arial",
        19,
        "bold"
    )
)

prediction_label.pack(
    pady=(5, 3)
)


# ============================================================
# CONFIDENCE
# ============================================================

confidence_label = Label(
    root,
    text="Confidence: ---",
    font=(
        "Arial",
        14
    )
)

confidence_label.pack(
    pady=2
)


# ============================================================
# TOP 3
# ============================================================

top3_label = Label(
    root,
    text="Top 3 Predictions\n\n---",
    font=(
        "Arial",
        12
    ),
    justify="left"
)

top3_label.pack(
    pady=5
)


# ============================================================
# WARNING
# ============================================================

warning_label = Label(
    root,
    text="",
    font=(
        "Arial",
        11
    )
)

warning_label.pack(
    pady=3
)


# ============================================================
# STATUS
# ============================================================

status_label = Label(
    root,
    text="Select an image to begin",
    font=(
        "Arial",
        10
    )
)

status_label.pack(
    pady=5
)


# ============================================================
# START
# ============================================================

print()
print("========================================")
print("SYSTEM READY")
print("========================================")
print(
    f"Model: {MODEL_PATH}"
)
print(
    f"Number of breeds: {len(CLASS_NAMES)}"
)
print(
    "Best validation accuracy: 70.31%"
)
print(
    "Waiting for image..."
)
print()

root.mainloop()
```

```python
import os
import warnings
import numpy as np
import cv2
import tensorflow as tf

from tkinter import Tk, Label, Button, filedialog, messagebox, Frame
from PIL import Image, ImageTk

warnings.filterwarnings("ignore")

# ============================================================
# SETTINGS
# ============================================================

MODEL_PATH = "Best_Cattle_Breed.h5"
DATA_DIR = "data"

IMAGE_SIZE = (224, 224)

# ============================================================
# LOAD MODEL
# ============================================================

print("Loading model...")

if not os.path.exists(MODEL_PATH):
    print("ERROR: Best_Cattle_Breed.h5 not found!")
    raise FileNotFoundError(
        "Best_Cattle_Breed.h5 was not found in the current folder."
    )

model = tf.keras.models.load_model(MODEL_PATH)

print("Model loaded successfully.")

# ============================================================
# LOAD CLASS NAMES
# ============================================================

if not os.path.exists(DATA_DIR):
    print("ERROR: data folder not found!")
    raise FileNotFoundError(
        "The 'data' folder was not found."
    )

CLASS_NAMES = sorted([
    folder for folder in os.listdir(DATA_DIR)
    if os.path.isdir(os.path.join(DATA_DIR, folder))
])

print("\nClasses detected:")
for i, name in enumerate(CLASS_NAMES):
    print(f"{i} -> {name}")

print(f"\nTotal classes: {len(CLASS_NAMES)}")

# ============================================================
# PREPROCESS IMAGE
# ============================================================

def preprocess_image(image_path):

    image = cv2.imread(image_path)

    if image is None:
        raise ValueError("Could not read the selected image.")

    # OpenCV BGR -> RGB
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Resize for model
    image = cv2.resize(image, IMAGE_SIZE)

    # Convert to float
    image = image.astype(np.float32)

    # EfficientNetV2 preprocessing
    image = tf.keras.applications.efficientnet_v2.preprocess_input(image)

    # Add batch dimension
    image = np.expand_dims(image, axis=0)

    return image


# ============================================================
# PREDICTION
# ============================================================

def predict_breed(image_path):

    processed_image = preprocess_image(image_path)

    prediction = model.predict(processed_image, verbose=0)

    # Handle different model output formats
    if prediction.ndim == 2:
        probabilities = prediction[0]
    else:
        probabilities = prediction.flatten()

    predicted_index = int(np.argmax(probabilities))

    confidence = float(probabilities[predicted_index]) * 100

    if predicted_index < len(CLASS_NAMES):
        breed = CLASS_NAMES[predicted_index]
    else:
        breed = f"Class {predicted_index}"

    return breed, confidence


# ============================================================
# GUI
# ============================================================

root = Tk()

root.title("Cattle Breed Classification System")

# ============================================================
# FULLSCREEN / MAXIMIZED WINDOW
# ============================================================

try:
    root.state("zoomed")
except:
    screen_width = root.winfo_screenwidth()
    screen_height = root.winfo_screenheight()

    root.geometry(
        f"{screen_width}x{screen_height}+0+0"
    )

root.resizable(True, True)

# Minimum size
root.minsize(900, 650)

# ============================================================
# VARIABLES
# ============================================================

selected_image_path = None
display_image = None

# ============================================================
# TITLE
# ============================================================

title_label = Label(
    root,
    text="CATTLE BREED CLASSIFICATION SYSTEM",
    font=("Arial", 26, "bold")
)

title_label.pack(pady=20)

subtitle_label = Label(
    root,
    text="Upload a cattle image to predict its breed",
    font=("Arial", 14)
)

subtitle_label.pack(pady=5)

# ============================================================
# MAIN FRAME
# ============================================================

main_frame = Frame(root)
main_frame.pack(
    fill="both",
    expand=True,
    padx=30,
    pady=20
)

# ============================================================
# IMAGE FRAME
# ============================================================

image_frame = Frame(
    main_frame,
    bd=2,
    relief="groove"
)

image_frame.pack(
    fill="both",
    expand=True,
    padx=20,
    pady=10
)

image_label = Label(
    image_frame,
    text="No image selected",
    font=("Arial", 18)
)

image_label.pack(
    fill="both",
    expand=True,
    padx=20,
    pady=20
)

# ============================================================
# RESULT
# ============================================================

result_label = Label(
    main_frame,
    text="Prediction will appear here",
    font=("Arial", 22, "bold")
)

result_label.pack(pady=15)

confidence_label = Label(
    main_frame,
    text="",
    font=("Arial", 16)
)

confidence_label.pack(pady=5)

# ============================================================
# SHOW IMAGE
# ============================================================

def show_image(image_path):

    global display_image

    try:

        original = Image.open(image_path)

        # Get available area
        image_frame.update_idletasks()

        max_width = image_frame.winfo_width() - 50
        max_height = image_frame.winfo_height() - 50

        # Safety values
        if max_width < 300:
            max_width = 800

        if max_height < 300:
            max_height = 500

        # Keep original aspect ratio
        original.thumbnail(
            (max_width, max_height),
            Image.Resampling.LANCZOS
        )

        display_image = ImageTk.PhotoImage(original)

        image_label.configure(
            image=display_image,
            text=""
        )

        image_label.image = display_image

    except Exception as e:

        messagebox.showerror(
            "Image Error",
            f"Could not display image:\n\n{e}"
        )


# ============================================================
# UPLOAD IMAGE
# ============================================================

def select_image():

    global selected_image_path

    file_path = filedialog.askopenfilename(
        title="Select Cattle Image",
        filetypes=[
            ("Image Files", "*.jpg *.jpeg *.png *.bmp *.webp"),
            ("JPG Files", "*.jpg"),
            ("PNG Files", "*.png"),
            ("All Files", "*.*")
        ]
    )

    if not file_path:
        return

    selected_image_path = file_path

    show_image(file_path)

    result_label.config(
        text="Image selected"
    )

    confidence_label.config(
        text="Click 'Predict Breed'"
    )


# ============================================================
# PREDICT BUTTON
# ============================================================

def predict_selected_image():

    if selected_image_path is None:

        messagebox.showwarning(
            "No Image",
            "Please select a cattle image first."
        )

        return

    try:

        result_label.config(
            text="Predicting..."
        )

        confidence_label.config(
            text=""
        )

        root.update()

        breed, confidence = predict_breed(
            selected_image_path
        )

        result_label.config(
            text=f"Predicted Breed: {breed}"
        )

        confidence_label.config(
            text=f"Confidence: {confidence:.2f}%"
        )

        print(
            f"\nPrediction: {breed}"
        )

        print(
            f"Confidence: {confidence:.2f}%"
        )

    except Exception as e:

        result_label.config(
            text="Prediction failed"
        )

        confidence_label.config(
            text=""
        )

        messagebox.showerror(
            "Prediction Error",
            f"Something went wrong:\n\n{e}"
        )


# ============================================================
# BUTTON FRAME
# ============================================================

button_frame = Frame(root)

button_frame.pack(
    pady=20
)

# ============================================================
# SELECT IMAGE BUTTON
# ============================================================

select_button = Button(
    button_frame,
    text="SELECT CATTLE IMAGE",
    command=select_image,
    font=("Arial", 14, "bold"),
    padx=25,
    pady=12
)

select_button.pack(
    side="left",
    padx=15
)

# ============================================================
# PREDICT BUTTON
# ============================================================

predict_button = Button(
    button_frame,
    text="PREDICT BREED",
    command=predict_selected_image,
    font=("Arial", 14, "bold"),
    padx=25,
    pady=12
)

predict_button.pack(
    side="left",
    padx=15
)

# ============================================================
# EXIT BUTTON
# ============================================================

exit_button = Button(
    button_frame,
    text="EXIT",
    command=root.destroy,
    font=("Arial", 14, "bold"),
    padx=25,
    pady=12
)

exit_button.pack(
    side="left",
    padx=15
)

# ============================================================
# START APPLICATION
# ============================================================

print("\n==========================================")
print(" CATTLE BREED CLASSIFICATION SYSTEM")
print("==========================================")
print("GUI started successfully.")
print("Select an image to begin.")

root.mainloop()
```
