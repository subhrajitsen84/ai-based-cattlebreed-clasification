```python
import os
import numpy as np
import tensorflow as tf
import cv2

MODEL_PATH = "Best_Cattle_Breed_Improved.h5"
IMAGE_SIZE = (224, 224)

CLASS_NAMES = [
    "Alambadi", "Amritmahal", "Ayrshire", "Banni", "Bargur", "Bhadawari",
    "Brown_Swiss", "Dangi", "Deoni", "Gir", "Guernsey", "Hallikar", "Hariana",
    "Holstein_Friesian", "Jaffrabadi", "Jersey", "Kangayam", "Kankrej",
    "Kasargod", "Kenkatha", "Kherigarh", "Khillari", "Krishna_Valley",
    "Malnad_gidda", "Mehsana", "Murrah", "Nagori", "Nagpuri", "Nili_Ravi",
    "Nimari", "Ongole", "Pulikulam", "Rathi", "Red_Dane", "Red_Sindhi",
    "Sahiwal", "Surti", "Tharparkar", "Toda", "Umblachery", "Vechur"
]

# Check model
if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError(f"Model not found: {MODEL_PATH}")

# Load breed classification model
model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

# Check output classes
if model.output_shape[-1] != len(CLASS_NAMES):
    raise ValueError(
        f"Model output has {model.output_shape[-1]} classes, "
        f"but CLASS_NAMES has {len(CLASS_NAMES)} classes."
    )

# Pretrained model used to check whether image contains cattle
cow_model = tf.keras.applications.MobileNetV2(
    weights="imagenet"
)


def predict(path):

    # -----------------------------
    # LOAD IMAGE
    # -----------------------------
    img = cv2.imread(path)

    if img is None:
        raise ValueError("Cannot read image")

    # -----------------------------
    # CATTLE / NON-CATTLE CHECK
    # -----------------------------
    check = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    check = cv2.resize(
        check,
        IMAGE_SIZE
    ).astype(np.float32)

    check = tf.keras.applications.mobilenet_v2.preprocess_input(
        check
    )

    check_pred = cow_model.predict(
        np.expand_dims(check, 0),
        verbose=0
    )

    labels = [
        x[1].lower()
        for x in tf.keras.applications.mobilenet_v2.decode_predictions(
            check_pred,
            top=5
        )[0]
    ]

    cattle_words = [
        "cow",
        "ox",
        "water_buffalo",
        "buffalo"
    ]

    # Reject non-cattle images
    if not any(
        word in label
        for label in labels
        for word in cattle_words
    ):
        return None

    # -----------------------------
    # BREED CLASSIFICATION
    # -----------------------------
    img = cv2.cvtColor(
        img,
        cv2.COLOR_BGR2RGB
    )

    img = cv2.resize(
        img,
        IMAGE_SIZE
    ).astype(np.float32)

    img = tf.keras.applications.efficientnet_v2.preprocess_input(
        img
    )

    pred = model.predict(
        np.expand_dims(img, 0),
        verbose=0
    )[0]

    # Get top 3 predictions
    ids = np.argsort(pred)[-3:][::-1]

    results = [
        (CLASS_NAMES[i], float(pred[i] * 100))
        for i in ids
    ]

    return results
```
